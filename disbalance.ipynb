```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import re
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModel,
    get_linear_schedule_with_warmup,
)
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from copy import deepcopy

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME  = "roberta-base"
MAX_LEN       = 128
BATCH_SIZE    = 32
EPOCHS        = 10
LEARNING_RATE = 2e-5
WARMUP_RATIO  = 0.1
WEIGHT_DECAY  = 0.01
PATIENCE      = 3
SEED          = 42

emotion_classes = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]

OUTPUT_DIR = "loss_balancing_experiments"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# ВОСПРОИЗВОДИМОСТЬ
# ============================================================

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

# ============================================================
# ОЧИСТКА ТЕКСТА
# ============================================================

def clean_platform(text):
    text = str(text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http[s]?://\S+', '', text)
    text = re.sub(r'www\.\S+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def prepare_platform(df_orig):
    df = df_orig.copy()
    df['tweet_text'] = df['tweet_text'].apply(clean_platform)
    df = df[df['tweet_text'].str.strip().str.len() > 0]
    return df.reset_index(drop=True)

# ============================================================
# DATASET
# ============================================================

class TweetDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts     = texts.reset_index(drop=True)
        self.labels    = labels.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            str(self.texts[idx]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_attention_mask=True,
            return_tensors="pt",
        )
        return {
            "input_ids":      encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels":         torch.tensor(int(self.labels[idx]), dtype=torch.long),
        }

# ============================================================
# МОДЕЛЬ — RoBERTa
# ============================================================

class RobertaEmotionModel(nn.Module):
    def __init__(self, model_name, num_labels):
        super().__init__()
        self.backbone   = AutoModel.from_pretrained(model_name)
        self.dropout    = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.backbone.config.hidden_size, num_labels)

    def forward(self, input_ids, attention_mask):
        out = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0]
        return self.classifier(self.dropout(cls))

# ============================================================
# АДАПТИРОВАННЫЕ ФУНКЦИИ ПОТЕРЬ
# ============================================================

class FocalLoss(nn.Module):
    """
    Focal Loss: FL(p_t) = -alpha_t * (1 - p_t)^gamma * log(p_t)
    Снижает вес лёгких примеров, фокусируя обучение на сложных.
    Lin et al. (2017).
    """
    def __init__(self, gamma: float = 2.0, weight: torch.Tensor = None,
                 reduction: str = "mean"):
        super().__init__()
        self.gamma     = gamma
        self.weight    = weight    # опциональные per-class веса
        self.reduction = reduction

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        # log_softmax для численной стабильности
        log_prob = F.log_softmax(logits, dim=-1)
        prob     = log_prob.exp()

        # Собираем вероятность истинного класса
        log_pt = log_prob.gather(1, targets.unsqueeze(1)).squeeze(1)
        pt     = prob.gather(1, targets.unsqueeze(1)).squeeze(1)

        focal_factor = (1.0 - pt) ** self.gamma
        loss         = -focal_factor * log_pt

        # Применяем per-class веса, если заданы
        if self.weight is not None:
            w    = self.weight.to(logits.device)
            loss = loss * w[targets]

        if self.reduction == "mean":
            return loss.mean()
        elif self.reduction == "sum":
            return loss.sum()
        return loss


class LabelSmoothingCrossEntropy(nn.Module):
    """
    Label Smoothing Cross-Entropy.
    Смягчает уверенность модели: истинный класс получает (1 - eps),
    остальные — eps / (C - 1).
    Улучшает калибровку, особенно на редких классах.
    """
    def __init__(self, smoothing: float = 0.1, weight: torch.Tensor = None):
        super().__init__()
        self.smoothing = smoothing
        self.weight    = weight

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        n_classes  = logits.size(-1)
        log_prob   = F.log_softmax(logits, dim=-1)

        # Сглаженное распределение меток
        with torch.no_grad():
            smooth_dist = torch.full_like(log_prob, self.smoothing / (n_classes - 1))
            smooth_dist.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)

        loss = -(smooth_dist * log_prob).sum(dim=-1)

        if self.weight is not None:
            w    = self.weight.to(logits.device)
            loss = loss * w[targets]

        return loss.mean()


class ClassBalancedLoss(nn.Module):
    """
    Class-Balanced Loss (Cui et al., Google Brain 2019).
    Вес класса: (1 - beta) / (1 - beta^n_i),
    где n_i — число примеров класса i, beta = (N-1)/N.
    Учитывает «эффективное число примеров» вместо сырых частот.
    """
    def __init__(self, samples_per_class: list, beta: float = None,
                 loss_type: str = "softmax"):
        super().__init__()
        if beta is None:
            N = sum(samples_per_class)
            beta = (N - 1) / N

        effective_num = [1.0 - beta ** n for n in samples_per_class]
        weights       = [(1.0 - beta) / en for en in effective_num]
        # Нормируем, чтобы сумма = num_classes
        total = sum(weights)
        n_cls = len(samples_per_class)
        weights = [w / total * n_cls for w in weights]

        self.register_buffer("weights", torch.tensor(weights, dtype=torch.float32))
        self.loss_type = loss_type

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        return F.cross_entropy(logits, targets, weight=self.weights.to(logits.device))


def build_class_weights_standard(samples_per_class: list,
                                   device: torch.device) -> torch.Tensor:
    """Стандартные веса 1/freq, нормированные."""
    freqs   = np.array(samples_per_class, dtype=float)
    weights = 1.0 / freqs
    weights = weights / weights.sum() * len(samples_per_class)
    return torch.tensor(weights, dtype=torch.float32).to(device)


def build_class_weights_sqrt(samples_per_class: list,
                              device: torch.device) -> torch.Tensor:
    """Кастомные веса 1/sqrt(freq), нормированные."""
    freqs   = np.array(samples_per_class, dtype=float)
    weights = 1.0 / np.sqrt(freqs)
    weights = weights / weights.sum() * len(samples_per_class)
    return torch.tensor(weights, dtype=torch.float32).to(device)


# ============================================================
# TRAIN / EVAL
# ============================================================

def train_epoch(model, loader, optimizer, scheduler, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []

    for batch in tqdm(loader, desc="  train", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        optimizer.zero_grad()
        logits = model(ids, mask)
        loss   = criterion(logits, labs)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        all_preds.extend(logits.argmax(1).cpu().numpy())
        all_labels.extend(labs.cpu().numpy())

    return total_loss / len(loader), f1_score(all_labels, all_preds, average="macro")


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = 0
    all_preds, all_labels, all_probs = [], [], []
    ce = nn.CrossEntropyLoss()

    for batch in tqdm(loader, desc="  eval", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        logits = model(ids, mask)
        total_loss += ce(logits, labs).item()
        probs = torch.softmax(logits, 1).cpu().numpy()

        all_preds.extend(probs.argmax(1))
        all_labels.extend(labs.cpu().numpy())
        all_probs.extend(probs)

    avg_loss = total_loss / len(loader)
    f1_mac   = f1_score(all_labels, all_preds, average="macro")
    return avg_loss, f1_mac, np.array(all_preds), np.array(all_labels), np.array(all_probs)

# ============================================================
# УНИВЕРСАЛЬНЫЙ ЭКСПЕРИМЕНТ
# ============================================================

def run_experiment(df_prepared, display_name, criterion_fn,
                   tokenizer, X_train, X_val, X_test,
                   y_train, y_val, y_test, le):
    """
    criterion_fn — callable(samples_per_class) -> nn.Module
                   принимает список числа примеров по классам,
                   возвращает готовый criterion.
    """
    print(f"\n{'='*60}")
    print(f"  {display_name}")
    print(f"{'='*60}")

    set_seed(SEED)
    num_labels = len(emotion_classes)

    # Считаем количество примеров по классам в train
    samples_per_class = [
        int((y_train == i).sum()) for i in range(num_labels)
    ]

    print(f"\n  Train class distribution:")
    for i, cls in enumerate(emotion_classes):
        print(f"    {cls:<12}: {samples_per_class[i]:>5}")

    criterion = criterion_fn(samples_per_class)

    tr_loader = DataLoader(
        TweetDataset(X_train, y_train, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE, shuffle=True)
    va_loader = DataLoader(
        TweetDataset(X_val, y_val, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)
    te_loader = DataLoader(
        TweetDataset(X_test, y_test, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)

    set_seed(SEED)
    model = RobertaEmotionModel(ROBERTA_NAME, num_labels).to(device)
    optimizer = torch.optim.AdamW(
        model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = len(tr_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(
        optimizer, int(total_steps * WARMUP_RATIO), total_steps)

    best_f1, patience_cnt, best_state = 0.0, 0, None
    history = {"train_loss": [], "val_loss": [], "train_f1": [], "val_f1": []}

    for ep in range(1, EPOCHS + 1):
        tr_loss, tr_f1 = train_epoch(model, tr_loader, optimizer, scheduler, criterion)
        va_loss, va_f1, _, _, _ = evaluate(model, va_loader)

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(va_loss)
        history["train_f1"].append(tr_f1)
        history["val_f1"].append(va_f1)

        print(f"  Ep {ep}/{EPOCHS}  "
              f"train_loss={tr_loss:.4f} train_f1={tr_f1:.4f}  "
              f"val_loss={va_loss:.4f} val_f1={va_f1:.4f}", end="")

        if va_f1 > best_f1:
            best_f1, patience_cnt = va_f1, 0
            best_state = deepcopy(model.state_dict())
            print("  ✓ best")
        else:
            patience_cnt += 1
            print(f"  ⏳ {patience_cnt}/{PATIENCE}")

        if patience_cnt >= PATIENCE:
            print(f"  Early stop at epoch {ep}")
            break

    model.load_state_dict(best_state)
    te_loss, te_f1, te_preds, te_labels, te_probs = evaluate(model, te_loader)

    target_names = le.inverse_transform(range(num_labels))
    rep     = classification_report(te_labels, te_preds,
                                    target_names=target_names, digits=4,
                                    output_dict=True)
    rep_str = classification_report(te_labels, te_preds,
                                    target_names=target_names, digits=4)

    print(f"\n  Test loss={te_loss:.4f}  Macro F1={te_f1:.4f}")
    print(rep_str)

    result = {
        "variant":               display_name,
        "train_size":            len(X_train),
        "val_size":              len(X_val),
        "test_size":             len(X_test),
        "epochs_trained":        len(history["train_loss"]),
        "best_val_f1":           best_f1,
        "test_loss":             te_loss,
        "test_accuracy":         rep["accuracy"],
        "test_macro_precision":  rep["macro avg"]["precision"],
        "test_macro_recall":     rep["macro avg"]["recall"],
        "test_macro_f1":         rep["macro avg"]["f1-score"],
        "test_weighted_f1":      rep["weighted avg"]["f1-score"],
        "history":               history,
        "report_dict":           rep,
        "test_preds":            te_preds,
        "test_labels":           te_labels,
        "label_encoder":         le,
    }
    for cls in emotion_classes:
        result[f"f1_{cls}"] = rep[cls]["f1-score"] if cls in rep else 0.0

    return result

# ============================================================
# ЗАГРУЗКА ДАННЫХ И ЕДИНЫЙ СПЛИТ
# ============================================================

df_original = pd.read_csv("dataset_final_verified_v2_neutral.csv")
df_clean    = prepare_platform(df_original)

df_exp = df_clean[
    df_clean["final_label"].isin(emotion_classes)
].reset_index(drop=True)

le = LabelEncoder()
le.fit(emotion_classes)
df_exp["labels"] = le.transform(df_exp["final_label"])

print(f"\nDataset size: {len(df_exp)}")
dist = df_exp["final_label"].value_counts()
for cls in emotion_classes:
    n   = dist.get(cls, 0)
    pct = n / len(df_exp) * 100
    print(f"  {cls:<12}: {n:>5}  ({pct:.1f}%)")

# Единый сплит для всех экспериментов — честное сравнение
X_tv, X_test, y_tv, y_test = train_test_split(
    df_exp["tweet_text"], df_exp["labels"],
    test_size=0.15, random_state=SEED, stratify=df_exp["labels"])
X_train, X_val, y_train, y_val = train_test_split(
    X_tv, y_tv,
    test_size=0.15, random_state=SEED, stratify=y_tv)

print(f"\nTrain={len(X_train)}  Val={len(X_val)}  Test={len(X_test)}")

tokenizer = AutoTokenizer.from_pretrained(ROBERTA_NAME)

# ============================================================
# ОПРЕДЕЛЕНИЕ ЭКСПЕРИМЕНТОВ
# ============================================================
# criterion_fn: callable(samples_per_class: list) -> nn.Module

experiments = [
    # 1. Baseline — без балансировки
    {
        "display_name": "1. Baseline (no balancing)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(),
    },
    # 2. Class-Weighted CE — стандартные веса 1/freq
    {
        "display_name": "2. Weighted CE (1/freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(
            weight=build_class_weights_standard(spc, device)
        ),
    },
    # 3. Class-Weighted CE — кастомные веса 1/sqrt(freq)
    {
        "display_name": "3. Weighted CE (1/sqrt freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(
            weight=build_class_weights_sqrt(spc, device)
        ),
    },
    # 4. Focal Loss (gamma=2, без доп. весов)
    {
        "display_name": "4. Focal Loss (gamma=2)",
        "criterion_fn": lambda spc: FocalLoss(gamma=2.0),
    },
    # 5. Focal Loss + Class Weights 1/sqrt(freq)
    {
        "display_name": "5. Focal Loss + Weighted (1/sqrt freq)",
        "criterion_fn": lambda spc: FocalLoss(
            gamma=2.0,
            weight=build_class_weights_sqrt(spc, device)
        ),
    },
    # 6. Label Smoothing (без доп. весов)
    {
        "display_name": "6. Label Smoothing (eps=0.1)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(smoothing=0.1),
    },
    # 7. Label Smoothing + Class Weights 1/sqrt(freq)
    {
        "display_name": "7. Label Smoothing + Weighted (1/sqrt freq)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(
            smoothing=0.1,
            weight=build_class_weights_sqrt(spc, device)
        ),
    },
    # 8. Class-Balanced Loss (Google Brain 2019)
    {
        "display_name": "8. Class-Balanced Loss (beta auto)",
        "criterion_fn": lambda spc: ClassBalancedLoss(samples_per_class=spc),
    },
]

# ============================================================
# ЗАПУСК ВСЕХ ЭКСПЕРИМЕНТОВ
# ============================================================

all_results = []
for exp in experiments:
    res = run_experiment(
        df_prepared  = df_exp,
        display_name = exp["display_name"],
        criterion_fn = exp["criterion_fn"],
        tokenizer    = tokenizer,
        X_train=X_train, X_val=X_val, X_test=X_test,
        y_train=y_train, y_val=y_val, y_test=y_test,
        le=le,
    )
    all_results.append(res)

# ============================================================
# СТИЛИСТИКА ГРАФИКОВ
# ============================================================

plt.rcParams['font.family'] = 'DejaVu Sans'

PROPS_BOX = dict(boxstyle='round,pad=0.6',
                 facecolor='#f5f5f5', edgecolor='#aaaaaa', alpha=0.9)

# Палитра для экспериментов
EXP_COLORS = [
    '#2c7bb6', '#1a9641', '#d7191c', '#fdae61',
    '#abd9e9', '#a6d96a', '#762a83', '#e66101',
]

# ============================================================
# ГРАФИКИ ПО КАЖДОМУ ЭКСПЕРИМЕНТУ
# ============================================================

def plot_training_curves(result, output_dir):
    ep_range = range(1, len(result["history"]["train_f1"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.patch.set_facecolor('white')

    for ax in axes:
        ax.set_facecolor('white')
        ax.grid(True, linestyle='--', alpha=0.4, zorder=1)
        ax.set_axisbelow(True)

    axes[0].plot(ep_range, result["history"]["train_f1"],
                 "o-", color="steelblue", linewidth=2.5, markersize=6,
                 label="Train Macro F1", zorder=3)
    axes[0].plot(ep_range, result["history"]["val_f1"],
                 "s-", color="#e74c3c", linewidth=2.5, markersize=6,
                 label="Val Macro F1", zorder=3)
    axes[0].set_title("Macro F1 per Epoch", fontsize=13, fontweight="bold", pad=12)
    axes[0].set_xlabel("Epoch", fontsize=12)
    axes[0].set_ylabel("Macro F1", fontsize=12)
    axes[0].legend(fontsize=11)

    axes[1].plot(ep_range, result["history"]["train_loss"],
                 "o-", color="steelblue", linewidth=2.5, markersize=6,
                 label="Train Loss", zorder=3)
    axes[1].plot(ep_range, result["history"]["val_loss"],
                 "s-", color="#e74c3c", linewidth=2.5, markersize=6,
                 label="Val Loss", zorder=3)
    axes[1].set_title("Loss per Epoch", fontsize=13, fontweight="bold", pad=12)
    axes[1].set_xlabel("Epoch", fontsize=12)
    axes[1].set_ylabel("Loss", fontsize=12)
    axes[1].legend(fontsize=11)

    stats_text = (
        f"TRAINING STATISTICS:\n"
        f"  • Epochs trained:    {result['epochs_trained']}\n"
        f"  • Best Val Macro F1: {result['best_val_f1']:.4f}\n"
        f"  • Train size:        {result['train_size']:,}\n"
        f"  • Val size:          {result['val_size']:,}"
    )
    axes[1].text(0.99, 0.03, stats_text,
                 transform=axes[1].transAxes, fontsize=9.5,
                 verticalalignment='bottom', horizontalalignment='right',
                 bbox=PROPS_BOX, family='monospace')

    plt.suptitle(result["variant"], fontsize=13, fontweight="bold", y=1.01)
    plt.tight_layout()

    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"curves_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


def plot_per_class_f1(result, output_dir):
    rep     = result["report_dict"]
    f1_vals = [rep[cls]["f1-score"] if cls in rep else 0 for cls in emotion_classes]

    fig, ax = plt.subplots(figsize=(11, 6))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')

    bars = ax.bar(emotion_classes, f1_vals, color="steelblue",
                  alpha=0.75, edgecolor='white', linewidth=0.8, zorder=2)
    for b, v in zip(bars, f1_vals):
        ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.012,
                f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

    ax.axhline(result["test_macro_f1"], color="#FFA500", linestyle="--",
               linewidth=2.2, zorder=3,
               label=f"Macro F1 = {result['test_macro_f1']:.4f}")
    ax.set_ylim(0, 1.05)
    ax.set_ylabel("F1-score", fontsize=12)
    ax.set_xlabel("Emotion Class", fontsize=12)
    ax.set_title(f"Per-Class F1 — {result['variant']}",
                 fontsize=12, fontweight="bold", pad=12)
    ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
    ax.set_axisbelow(True)
    ax.legend(fontsize=11)

    stats_text = (
        f"TEST STATISTICS:\n"
        f"  • Accuracy:    {result['test_accuracy']:.4f}\n"
        f"  • Macro P:     {result['test_macro_precision']:.4f}\n"
        f"  • Macro R:     {result['test_macro_recall']:.4f}\n"
        f"  • Macro F1:    {result['test_macro_f1']:.4f}\n"
        f"  • Weighted F1: {result['test_weighted_f1']:.4f}\n"
        f"  • Test size:   {result['test_size']:,}"
    )
    ax.text(0.99, 0.03, stats_text,
            transform=ax.transAxes, fontsize=9.5,
            verticalalignment='bottom', horizontalalignment='right',
            bbox=PROPS_BOX, family='monospace')

    plt.tight_layout()
    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"per_class_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


def plot_confusion_matrix(result, output_dir):
    fig, ax = plt.subplots(figsize=(9, 7))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')

    cm   = confusion_matrix(result["test_labels"], result["test_preds"])
    cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)

    sns.heatmap(cm_n, annot=True, fmt=".2f", cmap="Blues",
                xticklabels=emotion_classes,
                yticklabels=emotion_classes,
                ax=ax, linewidths=0.5, linecolor='white')

    ax.set_title(
        f"Normalized Confusion Matrix\n{result['variant']}\n"
        f"Macro F1 = {result['test_macro_f1']:.4f}  |  "
        f"Accuracy = {result['test_accuracy']:.4f}",
        fontsize=11, fontweight="bold", pad=12)
    ax.set_ylabel("True Label", fontsize=12)
    ax.set_xlabel("Predicted Label", fontsize=12)

    plt.tight_layout()
    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"cm_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


# Индивидуальные графики для каждого эксперимента
for res in all_results:
    plot_training_curves(res, OUTPUT_DIR)
    plot_per_class_f1(res, OUTPUT_DIR)
    plot_confusion_matrix(res, OUTPUT_DIR)

# ============================================================
# СРАВНИТЕЛЬНЫЙ ГРАФИК — MACRO F1 ПО ВСЕМ МЕТОДАМ
# ============================================================

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

short_names = [r["variant"] for r in all_results]
macro_f1s   = [r["test_macro_f1"] for r in all_results]
colors_bar  = EXP_COLORS[:len(all_results)]

bars = ax.bar(range(len(all_results)), macro_f1s, color=colors_bar,
              alpha=0.82, edgecolor='white', linewidth=0.8, zorder=2)

for i, (b, v) in enumerate(zip(bars, macro_f1s)):
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.003,
            f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

baseline_f1 = all_results[0]["test_macro_f1"]
ax.axhline(baseline_f1, color='black', linestyle='--', linewidth=1.5,
           zorder=3, label=f"Baseline = {baseline_f1:.4f}")

ax.set_xticks(range(len(all_results)))
ax.set_xticklabels([r["variant"] for r in all_results],
                   rotation=30, ha='right', fontsize=10)
ax.set_ylabel("Macro F1", fontsize=12, labelpad=8)
ax.set_title("Loss Function Comparison — Macro F1 (RoBERTa)",
             fontsize=13, fontweight="bold", pad=12)
ax.set_ylim(max(0, min(macro_f1s) - 0.05), min(1.0, max(macro_f1s) + 0.06))
ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
ax.set_axisbelow(True)
ax.legend(fontsize=11)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_macro_f1.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# СРАВНИТЕЛЬНЫЙ ГРАФИК — PER-CLASS F1 ПО ВСЕМ МЕТОДАМ
# ============================================================

n_exp = len(all_results)
x     = np.arange(len(emotion_classes))
w     = 0.10   # ширина столбца

fig, ax = plt.subplots(figsize=(16, 7))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

for i, (res, color) in enumerate(zip(all_results, EXP_COLORS)):
    f1s  = [res.get(f"f1_{c}", 0) for c in emotion_classes]
    offset = (i - n_exp / 2 + 0.5) * w
    bars = ax.bar(x + offset, f1s, w, label=res["variant"],
                  color=color, alpha=0.80, edgecolor='white', zorder=2)

ax.set_ylabel("F1-score", fontsize=12, labelpad=8)
ax.set_xlabel("Emotion Class", fontsize=12, labelpad=8)
ax.set_title("Per-Class F1 — All Loss Methods (RoBERTa)",
             fontsize=13, fontweight="bold", pad=12)
ax.set_xticks(x)
ax.set_xticklabels(emotion_classes, fontsize=12)
ax.set_ylim(0, 1.10)
ax.legend(fontsize=8, framealpha=0.9, edgecolor='#cccccc',
          loc='upper right', ncol=2)
ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_per_class_f1_all.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ТЕПЛОВАЯ КАРТА DELTA per-class F1 (vs Baseline)
# ============================================================

baseline_per_class = {cls: all_results[0].get(f"f1_{cls}", 0) for cls in emotion_classes}

delta_rows = []
for res in all_results[1:]:   # пропускаем baseline
    row = {"Method": res["variant"]}
    for cls in emotion_classes:
        row[cls] = round(res.get(f"f1_{cls}", 0) - baseline_per_class[cls], 4)
    delta_rows.append(row)

delta_df = pd.DataFrame(delta_rows).set_index("Method")

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white')

sns.heatmap(delta_df, annot=True, fmt=".4f", cmap="RdYlGn",
            center=0, linewidths=0.5, linecolor='white',
            ax=ax, cbar_kws={"label": "ΔF1 vs Baseline"})

ax.set_title("Per-Class F1 Delta vs Baseline (RoBERTa)\n"
             "Green = improvement, Red = degradation",
             fontsize=12, fontweight="bold", pad=12)
ax.set_xlabel("Emotion Class", fontsize=12)
ax.set_ylabel("Loss Method", fontsize=12)
ax.tick_params(axis='x', labelsize=11)
ax.tick_params(axis='y', labelsize=10)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "delta_per_class_f1_heatmap.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ИТОГОВЫЕ ТАБЛИЦЫ
# ============================================================

print(f"\n{'='*100}")
print("  LOSS FUNCTION COMPARISON — RoBERTa")
print(f"{'='*100}\n")

rows = []
for r in all_results:
    rows.append({
        "Method":       r["variant"],
        "Epochs":       r["epochs_trained"],
        "Accuracy":     round(r["test_accuracy"], 4),
        "Macro P":      round(r["test_macro_precision"], 4),
        "Macro R":      round(r["test_macro_recall"], 4),
        "Macro F1":     round(r["test_macro_f1"], 4),
        "Weighted F1":  round(r["test_weighted_f1"], 4),
        "Best Val F1":  round(r["best_val_f1"], 4),
        "Test Loss":    round(r["test_loss"], 4),
    })

summary_df = pd.DataFrame(rows).set_index("Method")
print(summary_df.to_string())

# Дельта к baseline
print(f"\n{'='*100}")
print("  DELTA vs BASELINE (Macro F1)")
print(f"{'='*100}\n")

baseline_macro = all_results[0]["test_macro_f1"]
for r in all_results[1:]:
    delta = r["test_macro_f1"] - baseline_macro
    sign  = "+" if delta >= 0 else ""
    print(f"  {r['variant']:<45}  ΔMacro F1 = {sign}{delta:.4f}")

print(f"\n{'='*100}")
print("  PER-CLASS F1 TABLE")
print(f"{'='*100}\n")

pc_rows = []
for r in all_results:
    row = {"Method": r["variant"]}
    for cls in emotion_classes:
        row[cls] = round(r.get(f"f1_{cls}", 0), 4)
    pc_rows.append(row)

pc_df = pd.DataFrame(pc_rows).set_index("Method")
print(pc_df.to_string())

print(f"\n{'='*100}")
print("  PER-CLASS F1 DELTA vs BASELINE")
print(f"{'='*100}\n")
print(delta_df.to_string())

# ============================================================
# СОХРАНЕНИЕ CSV
# ============================================================

summary_df.to_csv(os.path.join(OUTPUT_DIR, "summary_metrics.csv"))
pc_df.to_csv(os.path.join(OUTPUT_DIR, "per_class_f1.csv"))
delta_df.to_csv(os.path.join(OUTPUT_DIR, "delta_per_class_f1.csv"))

# Лучший метод по Macro F1
best_idx = int(np.argmax([r["test_macro_f1"] for r in all_results]))
print(f"\nBest method:  {all_results[best_idx]['variant']}")
print(f"  Macro F1  = {all_results[best_idx]['test_macro_f1']:.4f}")
print(f"  Accuracy  = {all_results[best_idx]['test_accuracy']:.4f}")
print(f"\nAll files saved to: {OUTPUT_DIR}/")
```

```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import re
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoTokenizer,
    AutoModel,
    get_linear_schedule_with_warmup,
)
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from copy import deepcopy

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME  = "roberta-base"
MAX_LEN       = 128
BATCH_SIZE    = 32
EPOCHS        = 10
LEARNING_RATE = 2e-5
WARMUP_RATIO  = 0.1
WEIGHT_DECAY  = 0.01
PATIENCE      = 3
SEED          = 42

emotion_classes = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]

OUTPUT_DIR = "loss_balancing_experiments"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# ВОСПРОИЗВОДИМОСТЬ
# ============================================================

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

# ============================================================
# ОЧИСТКА ТЕКСТА
# ============================================================

def clean_platform(text):
    text = str(text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http[s]?://\S+', '', text)
    text = re.sub(r'www\.\S+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def prepare_platform(df_orig):
    df = df_orig.copy()
    df['tweet_text'] = df['tweet_text'].apply(clean_platform)
    df = df[df['tweet_text'].str.strip().str.len() > 0]
    return df.reset_index(drop=True)

# ============================================================
# DATASET
# ============================================================

class TweetDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts     = texts.reset_index(drop=True)
        self.labels    = labels.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            str(self.texts[idx]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_attention_mask=True,
            return_tensors="pt",
        )
        return {
            "input_ids":      encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels":         torch.tensor(int(self.labels[idx]), dtype=torch.long),
        }

# ============================================================
# МОДЕЛЬ — RoBERTa
# ============================================================

class RobertaEmotionModel(nn.Module):
    def __init__(self, model_name, num_labels):
        super().__init__()
        self.backbone   = AutoModel.from_pretrained(model_name)
        self.dropout    = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.backbone.config.hidden_size, num_labels)

    def forward(self, input_ids, attention_mask):
        out = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0]
        return self.classifier(self.dropout(cls))

# ============================================================
# АДАПТИРОВАННЫЕ ФУНКЦИИ ПОТЕРЬ
# ============================================================

class FocalLoss(nn.Module):
    """
    Focal Loss: FL(p_t) = -alpha_t * (1 - p_t)^gamma * log(p_t)
    Снижает вес лёгких примеров, фокусируя обучение на сложных.
    Lin et al. (2017).
    """
    def __init__(self, gamma: float = 2.0, weight: torch.Tensor = None,
                 reduction: str = "mean"):
        super().__init__()
        self.gamma     = gamma
        self.weight    = weight    # опциональные per-class веса
        self.reduction = reduction

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        # log_softmax для численной стабильности
        log_prob = F.log_softmax(logits, dim=-1)
        prob     = log_prob.exp()

        # Собираем вероятность истинного класса
        log_pt = log_prob.gather(1, targets.unsqueeze(1)).squeeze(1)
        pt     = prob.gather(1, targets.unsqueeze(1)).squeeze(1)

        focal_factor = (1.0 - pt) ** self.gamma
        loss         = -focal_factor * log_pt

        # Применяем per-class веса, если заданы
        if self.weight is not None:
            w    = self.weight.to(logits.device)
            loss = loss * w[targets]

        if self.reduction == "mean":
            return loss.mean()
        elif self.reduction == "sum":
            return loss.sum()
        return loss


class LabelSmoothingCrossEntropy(nn.Module):
    """
    Label Smoothing Cross-Entropy.
    Смягчает уверенность модели: истинный класс получает (1 - eps),
    остальные — eps / (C - 1).
    Улучшает калибровку, особенно на редких классах.
    """
    def __init__(self, smoothing: float = 0.1, weight: torch.Tensor = None):
        super().__init__()
        self.smoothing = smoothing
        self.weight    = weight

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        n_classes  = logits.size(-1)
        log_prob   = F.log_softmax(logits, dim=-1)

        # Сглаженное распределение меток
        with torch.no_grad():
            smooth_dist = torch.full_like(log_prob, self.smoothing / (n_classes - 1))
            smooth_dist.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)

        loss = -(smooth_dist * log_prob).sum(dim=-1)

        if self.weight is not None:
            w    = self.weight.to(logits.device)
            loss = loss * w[targets]

        return loss.mean()


class ClassBalancedLoss(nn.Module):
    """
    Class-Balanced Loss (Cui et al., Google Brain 2019).
    Вес класса: (1 - beta) / (1 - beta^n_i),
    где n_i — число примеров класса i, beta = (N-1)/N.
    Учитывает «эффективное число примеров» вместо сырых частот.
    """
    def __init__(self, samples_per_class: list, beta: float = None,
                 loss_type: str = "softmax"):
        super().__init__()
        if beta is None:
            N = sum(samples_per_class)
            beta = (N - 1) / N

        effective_num = [1.0 - beta ** n for n in samples_per_class]
        weights       = [(1.0 - beta) / en for en in effective_num]
        # Нормируем, чтобы сумма = num_classes
        total = sum(weights)
        n_cls = len(samples_per_class)
        weights = [w / total * n_cls for w in weights]

        self.register_buffer("weights", torch.tensor(weights, dtype=torch.float32))
        self.loss_type = loss_type

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        return F.cross_entropy(logits, targets, weight=self.weights.to(logits.device))


def build_class_weights_standard(samples_per_class: list,
                                   device: torch.device) -> torch.Tensor:
    """Стандартные веса 1/freq, нормированные."""
    freqs   = np.array(samples_per_class, dtype=float)
    weights = 1.0 / freqs
    weights = weights / weights.sum() * len(samples_per_class)
    return torch.tensor(weights, dtype=torch.float32).to(device)


def build_class_weights_sqrt(samples_per_class: list,
                              device: torch.device) -> torch.Tensor:
    """Кастомные веса 1/sqrt(freq), нормированные."""
    freqs   = np.array(samples_per_class, dtype=float)
    weights = 1.0 / np.sqrt(freqs)
    weights = weights / weights.sum() * len(samples_per_class)
    return torch.tensor(weights, dtype=torch.float32).to(device)


# ============================================================
# TRAIN / EVAL
# ============================================================

def train_epoch(model, loader, optimizer, scheduler, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []

    for batch in tqdm(loader, desc="  train", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        optimizer.zero_grad()
        logits = model(ids, mask)
        loss   = criterion(logits, labs)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        all_preds.extend(logits.argmax(1).cpu().numpy())
        all_labels.extend(labs.cpu().numpy())

    return total_loss / len(loader), f1_score(all_labels, all_preds, average="macro")


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = 0
    all_preds, all_labels, all_probs = [], [], []
    ce = nn.CrossEntropyLoss()

    for batch in tqdm(loader, desc="  eval", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        logits = model(ids, mask)
        total_loss += ce(logits, labs).item()
        probs = torch.softmax(logits, 1).cpu().numpy()

        all_preds.extend(probs.argmax(1))
        all_labels.extend(labs.cpu().numpy())
        all_probs.extend(probs)

    avg_loss = total_loss / len(loader)
    f1_mac   = f1_score(all_labels, all_preds, average="macro")
    return avg_loss, f1_mac, np.array(all_preds), np.array(all_labels), np.array(all_probs)

# ============================================================
# УНИВЕРСАЛЬНЫЙ ЭКСПЕРИМЕНТ
# ============================================================

def run_experiment(df_prepared, display_name, criterion_fn,
                   tokenizer, X_train, X_val, X_test,
                   y_train, y_val, y_test, le):
    """
    criterion_fn — callable(samples_per_class) -> nn.Module
                   принимает список числа примеров по классам,
                   возвращает готовый criterion.
    """
    print(f"\n{'='*60}")
    print(f"  {display_name}")
    print(f"{'='*60}")

    set_seed(SEED)
    num_labels = len(emotion_classes)

    # Считаем количество примеров по классам в train
    samples_per_class = [
        int((y_train == i).sum()) for i in range(num_labels)
    ]

    print(f"\n  Train class distribution:")
    for i, cls in enumerate(emotion_classes):
        print(f"    {cls:<12}: {samples_per_class[i]:>5}")

    criterion = criterion_fn(samples_per_class)

    tr_loader = DataLoader(
        TweetDataset(X_train, y_train, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE, shuffle=True)
    va_loader = DataLoader(
        TweetDataset(X_val, y_val, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)
    te_loader = DataLoader(
        TweetDataset(X_test, y_test, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)

    set_seed(SEED)
    model = RobertaEmotionModel(ROBERTA_NAME, num_labels).to(device)
    optimizer = torch.optim.AdamW(
        model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = len(tr_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(
        optimizer, int(total_steps * WARMUP_RATIO), total_steps)

    best_f1, patience_cnt, best_state = 0.0, 0, None
    history = {"train_loss": [], "val_loss": [], "train_f1": [], "val_f1": []}

    for ep in range(1, EPOCHS + 1):
        tr_loss, tr_f1 = train_epoch(model, tr_loader, optimizer, scheduler, criterion)
        va_loss, va_f1, _, _, _ = evaluate(model, va_loader)

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(va_loss)
        history["train_f1"].append(tr_f1)
        history["val_f1"].append(va_f1)

        print(f"  Ep {ep}/{EPOCHS}  "
              f"train_loss={tr_loss:.4f} train_f1={tr_f1:.4f}  "
              f"val_loss={va_loss:.4f} val_f1={va_f1:.4f}", end="")

        if va_f1 > best_f1:
            best_f1, patience_cnt = va_f1, 0
            best_state = deepcopy(model.state_dict())
            print("  ✓ best")
        else:
            patience_cnt += 1
            print(f"  ⏳ {patience_cnt}/{PATIENCE}")

        if patience_cnt >= PATIENCE:
            print(f"  Early stop at epoch {ep}")
            break

    model.load_state_dict(best_state)
    te_loss, te_f1, te_preds, te_labels, te_probs = evaluate(model, te_loader)

    target_names = le.inverse_transform(range(num_labels))
    rep     = classification_report(te_labels, te_preds,
                                    target_names=target_names, digits=4,
                                    output_dict=True)
    rep_str = classification_report(te_labels, te_preds,
                                    target_names=target_names, digits=4)

    print(f"\n  Test loss={te_loss:.4f}  Macro F1={te_f1:.4f}")
    print(rep_str)

    result = {
        "variant":               display_name,
        "train_size":            len(X_train),
        "val_size":              len(X_val),
        "test_size":             len(X_test),
        "epochs_trained":        len(history["train_loss"]),
        "best_val_f1":           best_f1,
        "test_loss":             te_loss,
        "test_accuracy":         rep["accuracy"],
        "test_macro_precision":  rep["macro avg"]["precision"],
        "test_macro_recall":     rep["macro avg"]["recall"],
        "test_macro_f1":         rep["macro avg"]["f1-score"],
        "test_weighted_f1":      rep["weighted avg"]["f1-score"],
        "history":               history,
        "report_dict":           rep,
        "test_preds":            te_preds,
        "test_labels":           te_labels,
        "label_encoder":         le,
    }
    for cls in emotion_classes:
        result[f"f1_{cls}"] = rep[cls]["f1-score"] if cls in rep else 0.0

    return result

# ============================================================
# ЗАГРУЗКА ДАННЫХ И ЕДИНЫЙ СПЛИТ
# ============================================================

df_original = pd.read_csv("dataset_final_verified_v2_neutral.csv")
df_clean    = prepare_platform(df_original)

df_exp = df_clean[
    df_clean["final_label"].isin(emotion_classes)
].reset_index(drop=True)

le = LabelEncoder()
le.fit(emotion_classes)
df_exp["labels"] = le.transform(df_exp["final_label"])

print(f"\nDataset size: {len(df_exp)}")
dist = df_exp["final_label"].value_counts()
for cls in emotion_classes:
    n   = dist.get(cls, 0)
    pct = n / len(df_exp) * 100
    print(f"  {cls:<12}: {n:>5}  ({pct:.1f}%)")

# Единый сплит для всех экспериментов — честное сравнение
X_tv, X_test, y_tv, y_test = train_test_split(
    df_exp["tweet_text"], df_exp["labels"],
    test_size=0.15, random_state=SEED, stratify=df_exp["labels"])
X_train, X_val, y_train, y_val = train_test_split(
    X_tv, y_tv,
    test_size=0.15, random_state=SEED, stratify=y_tv)

print(f"\nTrain={len(X_train)}  Val={len(X_val)}  Test={len(X_test)}")

tokenizer = AutoTokenizer.from_pretrained(ROBERTA_NAME)

# ============================================================
# ОПРЕДЕЛЕНИЕ ЭКСПЕРИМЕНТОВ
# ============================================================
# criterion_fn: callable(samples_per_class: list) -> nn.Module

experiments = [
    # 1. Baseline — без балансировки
    {
        "display_name": "1. Baseline (no balancing)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(),
    },
    # 2. Class-Weighted CE — стандартные веса 1/freq
    {
        "display_name": "2. Weighted CE (1/freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(
            weight=build_class_weights_standard(spc, device)
        ),
    },
    # 3. Class-Weighted CE — кастомные веса 1/sqrt(freq)
    {
        "display_name": "3. Weighted CE (1/sqrt freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(
            weight=build_class_weights_sqrt(spc, device)
        ),
    },
    # 4. Focal Loss (gamma=2, без доп. весов)
    {
        "display_name": "4. Focal Loss (gamma=2)",
        "criterion_fn": lambda spc: FocalLoss(gamma=2.0),
    },
    # 5. Focal Loss + Class Weights 1/freq
    {
        "display_name": "5. Focal Loss + Weighted (1/freq)",
        "criterion_fn": lambda spc: FocalLoss(
            gamma=2.0,
            weight=build_class_weights_standard(spc, device)
        ),
    },
    # 6. Label Smoothing (без доп. весов)
    {
        "display_name": "6. Label Smoothing (eps=0.1)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(smoothing=0.1),
    },
    # 7. Label Smoothing + Class Weights 1/freq
    {
        "display_name": "7. Label Smoothing + Weighted (1/freq)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(
            smoothing=0.1,
            weight=build_class_weights_standard(spc, device)
        ),
    },
    # 8. Class-Balanced Loss (Google Brain 2019)
    {
        "display_name": "8. Class-Balanced Loss (beta auto)",
        "criterion_fn": lambda spc: ClassBalancedLoss(samples_per_class=spc),
    },
]

# ============================================================
# ЗАПУСК ВСЕХ ЭКСПЕРИМЕНТОВ
# ============================================================

all_results = []
for exp in experiments:
    res = run_experiment(
        df_prepared  = df_exp,
        display_name = exp["display_name"],
        criterion_fn = exp["criterion_fn"],
        tokenizer    = tokenizer,
        X_train=X_train, X_val=X_val, X_test=X_test,
        y_train=y_train, y_val=y_val, y_test=y_test,
        le=le,
    )
    all_results.append(res)

# ============================================================
# СТИЛИСТИКА ГРАФИКОВ
# ============================================================

plt.rcParams['font.family'] = 'DejaVu Sans'

PROPS_BOX = dict(boxstyle='round,pad=0.6',
                 facecolor='#f5f5f5', edgecolor='#aaaaaa', alpha=0.9)

# Палитра для экспериментов
EXP_COLORS = [
    '#2c7bb6', '#1a9641', '#d7191c', '#fdae61',
    '#abd9e9', '#a6d96a', '#762a83', '#e66101',
]

# ============================================================
# ГРАФИКИ ПО КАЖДОМУ ЭКСПЕРИМЕНТУ
# ============================================================

def plot_training_curves(result, output_dir):
    ep_range = range(1, len(result["history"]["train_f1"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.patch.set_facecolor('white')

    for ax in axes:
        ax.set_facecolor('white')
        ax.grid(True, linestyle='--', alpha=0.4, zorder=1)
        ax.set_axisbelow(True)

    axes[0].plot(ep_range, result["history"]["train_f1"],
                 "o-", color="steelblue", linewidth=2.5, markersize=6,
                 label="Train Macro F1", zorder=3)
    axes[0].plot(ep_range, result["history"]["val_f1"],
                 "s-", color="#e74c3c", linewidth=2.5, markersize=6,
                 label="Val Macro F1", zorder=3)
    axes[0].set_title("Macro F1 per Epoch", fontsize=13, fontweight="bold", pad=12)
    axes[0].set_xlabel("Epoch", fontsize=12)
    axes[0].set_ylabel("Macro F1", fontsize=12)
    axes[0].legend(fontsize=11)

    axes[1].plot(ep_range, result["history"]["train_loss"],
                 "o-", color="steelblue", linewidth=2.5, markersize=6,
                 label="Train Loss", zorder=3)
    axes[1].plot(ep_range, result["history"]["val_loss"],
                 "s-", color="#e74c3c", linewidth=2.5, markersize=6,
                 label="Val Loss", zorder=3)
    axes[1].set_title("Loss per Epoch", fontsize=13, fontweight="bold", pad=12)
    axes[1].set_xlabel("Epoch", fontsize=12)
    axes[1].set_ylabel("Loss", fontsize=12)
    axes[1].legend(fontsize=11)

    stats_text = (
        f"TRAINING STATISTICS:\n"
        f"  • Epochs trained:    {result['epochs_trained']}\n"
        f"  • Best Val Macro F1: {result['best_val_f1']:.4f}\n"
        f"  • Train size:        {result['train_size']:,}\n"
        f"  • Val size:          {result['val_size']:,}"
    )
    axes[1].text(0.99, 0.03, stats_text,
                 transform=axes[1].transAxes, fontsize=9.5,
                 verticalalignment='bottom', horizontalalignment='right',
                 bbox=PROPS_BOX, family='monospace')

    plt.suptitle(result["variant"], fontsize=13, fontweight="bold", y=1.01)
    plt.tight_layout()

    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"curves_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


def plot_per_class_f1(result, output_dir):
    rep     = result["report_dict"]
    f1_vals = [rep[cls]["f1-score"] if cls in rep else 0 for cls in emotion_classes]

    fig, ax = plt.subplots(figsize=(11, 6))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')

    bars = ax.bar(emotion_classes, f1_vals, color="steelblue",
                  alpha=0.75, edgecolor='white', linewidth=0.8, zorder=2)
    for b, v in zip(bars, f1_vals):
        ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.012,
                f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

    ax.axhline(result["test_macro_f1"], color="#FFA500", linestyle="--",
               linewidth=2.2, zorder=3,
               label=f"Macro F1 = {result['test_macro_f1']:.4f}")
    ax.set_ylim(0, 1.05)
    ax.set_ylabel("F1-score", fontsize=12)
    ax.set_xlabel("Emotion Class", fontsize=12)
    ax.set_title(f"Per-Class F1 — {result['variant']}",
                 fontsize=12, fontweight="bold", pad=12)
    ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
    ax.set_axisbelow(True)
    ax.legend(fontsize=11)

    stats_text = (
        f"TEST STATISTICS:\n"
        f"  • Accuracy:    {result['test_accuracy']:.4f}\n"
        f"  • Macro P:     {result['test_macro_precision']:.4f}\n"
        f"  • Macro R:     {result['test_macro_recall']:.4f}\n"
        f"  • Macro F1:    {result['test_macro_f1']:.4f}\n"
        f"  • Weighted F1: {result['test_weighted_f1']:.4f}\n"
        f"  • Test size:   {result['test_size']:,}"
    )
    ax.text(0.99, 0.03, stats_text,
            transform=ax.transAxes, fontsize=9.5,
            verticalalignment='bottom', horizontalalignment='right',
            bbox=PROPS_BOX, family='monospace')

    plt.tight_layout()
    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"per_class_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


def plot_confusion_matrix(result, output_dir):
    fig, ax = plt.subplots(figsize=(9, 7))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')

    cm   = confusion_matrix(result["test_labels"], result["test_preds"])
    cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)

    sns.heatmap(cm_n, annot=True, fmt=".2f", cmap="Blues",
                xticklabels=emotion_classes,
                yticklabels=emotion_classes,
                ax=ax, linewidths=0.5, linecolor='white')

    ax.set_title(
        f"Normalized Confusion Matrix\n{result['variant']}\n"
        f"Macro F1 = {result['test_macro_f1']:.4f}  |  "
        f"Accuracy = {result['test_accuracy']:.4f}",
        fontsize=11, fontweight="bold", pad=12)
    ax.set_ylabel("True Label", fontsize=12)
    ax.set_xlabel("Predicted Label", fontsize=12)

    plt.tight_layout()
    safe_name = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    fname = os.path.join(output_dir, f"cm_{safe_name}.png")
    plt.savefig(fname, dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()
    print(f"Saved: {fname}")


# Индивидуальные графики для каждого эксперимента
for res in all_results:
    plot_training_curves(res, OUTPUT_DIR)
    plot_per_class_f1(res, OUTPUT_DIR)
    plot_confusion_matrix(res, OUTPUT_DIR)

# ============================================================
# СРАВНИТЕЛЬНЫЙ ГРАФИК — MACRO F1 ПО ВСЕМ МЕТОДАМ
# ============================================================

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

short_names = [r["variant"] for r in all_results]
macro_f1s   = [r["test_macro_f1"] for r in all_results]
colors_bar  = EXP_COLORS[:len(all_results)]

bars = ax.bar(range(len(all_results)), macro_f1s, color=colors_bar,
              alpha=0.82, edgecolor='white', linewidth=0.8, zorder=2)

for i, (b, v) in enumerate(zip(bars, macro_f1s)):
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.003,
            f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

baseline_f1 = all_results[0]["test_macro_f1"]
ax.axhline(baseline_f1, color='black', linestyle='--', linewidth=1.5,
           zorder=3, label=f"Baseline = {baseline_f1:.4f}")

ax.set_xticks(range(len(all_results)))
ax.set_xticklabels([r["variant"] for r in all_results],
                   rotation=30, ha='right', fontsize=10)
ax.set_ylabel("Macro F1", fontsize=12, labelpad=8)
ax.set_title("Loss Function Comparison — Macro F1 (RoBERTa)",
             fontsize=13, fontweight="bold", pad=12)
ax.set_ylim(max(0, min(macro_f1s) - 0.05), min(1.0, max(macro_f1s) + 0.06))
ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
ax.set_axisbelow(True)
ax.legend(fontsize=11)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_macro_f1.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# СРАВНИТЕЛЬНЫЙ ГРАФИК — PER-CLASS F1 ПО ВСЕМ МЕТОДАМ
# ============================================================

n_exp = len(all_results)
x     = np.arange(len(emotion_classes))
w     = 0.10   # ширина столбца

fig, ax = plt.subplots(figsize=(16, 7))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

for i, (res, color) in enumerate(zip(all_results, EXP_COLORS)):
    f1s  = [res.get(f"f1_{c}", 0) for c in emotion_classes]
    offset = (i - n_exp / 2 + 0.5) * w
    bars = ax.bar(x + offset, f1s, w, label=res["variant"],
                  color=color, alpha=0.80, edgecolor='white', zorder=2)

ax.set_ylabel("F1-score", fontsize=12, labelpad=8)
ax.set_xlabel("Emotion Class", fontsize=12, labelpad=8)
ax.set_title("Per-Class F1 — All Loss Methods (RoBERTa)",
             fontsize=13, fontweight="bold", pad=12)
ax.set_xticks(x)
ax.set_xticklabels(emotion_classes, fontsize=12)
ax.set_ylim(0, 1.10)
ax.legend(fontsize=8, framealpha=0.9, edgecolor='#cccccc',
          loc='upper right', ncol=2)
ax.grid(True, axis='y', linestyle='--', alpha=0.4, zorder=1)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_per_class_f1_all.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ТЕПЛОВАЯ КАРТА DELTA per-class F1 (vs Baseline)
# ============================================================

baseline_per_class = {cls: all_results[0].get(f"f1_{cls}", 0) for cls in emotion_classes}

delta_rows = []
for res in all_results[1:]:   # пропускаем baseline
    row = {"Method": res["variant"]}
    for cls in emotion_classes:
        row[cls] = round(res.get(f"f1_{cls}", 0) - baseline_per_class[cls], 4)
    delta_rows.append(row)

delta_df = pd.DataFrame(delta_rows).set_index("Method")

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white')

sns.heatmap(delta_df, annot=True, fmt=".4f", cmap="RdYlGn",
            center=0, linewidths=0.5, linecolor='white',
            ax=ax, cbar_kws={"label": "ΔF1 vs Baseline"})

ax.set_title("Per-Class F1 Delta vs Baseline (RoBERTa)\n"
             "Green = improvement, Red = degradation",
             fontsize=12, fontweight="bold", pad=12)
ax.set_xlabel("Emotion Class", fontsize=12)
ax.set_ylabel("Loss Method", fontsize=12)
ax.tick_params(axis='x', labelsize=11)
ax.tick_params(axis='y', labelsize=10)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "delta_per_class_f1_heatmap.png"),
            dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ИТОГОВЫЕ ТАБЛИЦЫ
# ============================================================

print(f"\n{'='*100}")
print("  LOSS FUNCTION COMPARISON — RoBERTa")
print(f"{'='*100}\n")

rows = []
for r in all_results:
    rows.append({
        "Method":       r["variant"],
        "Epochs":       r["epochs_trained"],
        "Accuracy":     round(r["test_accuracy"], 4),
        "Macro P":      round(r["test_macro_precision"], 4),
        "Macro R":      round(r["test_macro_recall"], 4),
        "Macro F1":     round(r["test_macro_f1"], 4),
        "Weighted F1":  round(r["test_weighted_f1"], 4),
        "Best Val F1":  round(r["best_val_f1"], 4),
        "Test Loss":    round(r["test_loss"], 4),
    })

summary_df = pd.DataFrame(rows).set_index("Method")
print(summary_df.to_string())

# Дельта к baseline
print(f"\n{'='*100}")
print("  DELTA vs BASELINE (Macro F1)")
print(f"{'='*100}\n")

baseline_macro = all_results[0]["test_macro_f1"]
for r in all_results[1:]:
    delta = r["test_macro_f1"] - baseline_macro
    sign  = "+" if delta >= 0 else ""
    print(f"  {r['variant']:<45}  ΔMacro F1 = {sign}{delta:.4f}")

print(f"\n{'='*100}")
print("  PER-CLASS F1 TABLE")
print(f"{'='*100}\n")

pc_rows = []
for r in all_results:
    row = {"Method": r["variant"]}
    for cls in emotion_classes:
        row[cls] = round(r.get(f"f1_{cls}", 0), 4)
    pc_rows.append(row)

pc_df = pd.DataFrame(pc_rows).set_index("Method")
print(pc_df.to_string())

print(f"\n{'='*100}")
print("  PER-CLASS F1 DELTA vs BASELINE")
print(f"{'='*100}\n")
print(delta_df.to_string())

# ============================================================
# СОХРАНЕНИЕ CSV
# ============================================================

summary_df.to_csv(os.path.join(OUTPUT_DIR, "summary_metrics.csv"))
pc_df.to_csv(os.path.join(OUTPUT_DIR, "per_class_f1.csv"))
delta_df.to_csv(os.path.join(OUTPUT_DIR, "delta_per_class_f1.csv"))

# Лучший метод по Macro F1
best_idx = int(np.argmax([r["test_macro_f1"] for r in all_results]))
print(f"\nBest method:  {all_results[best_idx]['variant']}")
print(f"  Macro F1  = {all_results[best_idx]['test_macro_f1']:.4f}")
print(f"  Accuracy  = {all_results[best_idx]['test_accuracy']:.4f}")
print(f"\nAll files saved to: {OUTPUT_DIR}/")
```

```python
# ============================================================
# ЯЧЕЙКА 1: ПРЕДОБРАБОТКА, СПЛИТ, СОХРАНЕНИЕ, ЗАГРУЗКА
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME = "cirimus/modernbert-large-go-emotions"
MAX_LEN      = 128
BATCH_SIZE   = 32
SEED         = 42

emotion_classes = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]

# ФИКСИРОВАННЫЙ МАППИНГ — порядок ВСЕГДА как в emotion_classes
CLASS_TO_ID = {cls: i for i, cls in enumerate(emotion_classes)}
ID_TO_CLASS = {i: cls for i, cls in enumerate(emotion_classes)}

SPLITS_DIR = "splits_modernbert"
os.makedirs(SPLITS_DIR, exist_ok=True)

# ============================================================
# ВОСПРОИЗВОДИМОСТЬ
# ============================================================

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

set_seed(SEED)

# ============================================================
# ОЧИСТКА ТЕКСТА
# ============================================================

def clean_platform(text):
    text = str(text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http[s]?://\S+', '', text)
    text = re.sub(r'www\.\S+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def prepare_platform(df_orig):
    df = df_orig.copy()
    df['tweet_text'] = df['tweet_text'].apply(clean_platform)
    df = df[df['tweet_text'].str.strip().str.len() > 0]
    return df.reset_index(drop=True)

# ============================================================
# DATASET CLASS
# ============================================================

class TweetDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts     = texts.reset_index(drop=True)
        self.labels    = labels.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            str(self.texts[idx]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding="max_length",
            truncation=True,
            return_attention_mask=True,
            return_tensors="pt",
        )
        return {
            "input_ids":      encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels":         torch.tensor(int(self.labels[idx]), dtype=torch.long),
        }

# ============================================================
# ПОДГОТОВКА И СОХРАНЕНИЕ СПЛИТОВ
# ============================================================

def prepare_and_save_splits(
    csv_path: str,
    splits_dir: str = SPLITS_DIR,
    test_size: float = 0.15,
    val_size: float  = 0.15,
    seed: int        = SEED,
):
    from sklearn.model_selection import train_test_split

    print(f"Reading: {csv_path}")
    df_raw   = pd.read_csv(csv_path)
    df_clean = prepare_platform(df_raw)

    # Фильтр по известным классам
    df = df_clean[df_clean["final_label"].isin(emotion_classes)].reset_index(drop=True)
    print(f"Samples after filtering: {len(df)}")

    # ФИКСИРОВАННЫЙ маппинг (не LabelEncoder!)
    df["label_id"] = df["final_label"].map(CLASS_TO_ID)

    # Проверка что маппинг корректен
    assert df["label_id"].isna().sum() == 0, "Unmapped labels found!"

    # Распределение
    print("\nClass distribution (id → name → count):")
    for cls in emotion_classes:
        cid = CLASS_TO_ID[cls]
        n   = (df["label_id"] == cid).sum()
        pct = n / len(df) * 100
        print(f"  id={cid}  {cls:<12}: {n:>5}  ({pct:.1f}%)")

    # Сплит
    X_tv, X_test, y_tv, y_test = train_test_split(
        df[["tweet_text", "final_label", "label_id"]],
        df["label_id"],
        test_size=test_size, random_state=seed, stratify=df["label_id"],
    )
    X_train, X_val, y_train, y_val = train_test_split(
        X_tv, y_tv,
        test_size=val_size, random_state=seed, stratify=y_tv,
    )

    train_df = X_train.reset_index(drop=True)
    val_df   = X_val.reset_index(drop=True)
    test_df  = X_test.reset_index(drop=True)

    print(f"\nSplit sizes → Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

    # Проверка консистентности в train
    print("\nTrain verification (id → name → count):")
    for cls in emotion_classes:
        cid = CLASS_TO_ID[cls]
        n   = (train_df["label_id"] == cid).sum()
        print(f"  id={cid}  {cls:<12}: {n:>5}")

    # Сохранение
    train_df.to_csv(os.path.join(splits_dir, "train.csv"), index=False)
    val_df.to_csv(os.path.join(splits_dir, "val.csv"),     index=False)
    test_df.to_csv(os.path.join(splits_dir, "test.csv"),   index=False)

    # Маппинг меток — фиксированный порядок
    mapping_df = pd.DataFrame({
        "class_name": emotion_classes,
        "class_id":   [CLASS_TO_ID[c] for c in emotion_classes],
    })
    mapping_df.to_csv(os.path.join(splits_dir, "label_mapping.csv"), index=False)

    # Статистика
    info_lines = [
        f"Source file  : {csv_path}",
        f"Total samples: {len(df)}",
        f"Train size   : {len(train_df)}",
        f"Val size     : {len(val_df)}",
        f"Test size    : {len(test_df)}",
        f"Seed         : {seed}",
        "",
        "Label mapping:",
    ]
    for cls in emotion_classes:
        info_lines.append(f"  {CLASS_TO_ID[cls]} → {cls}")

    info_lines.append("\nClass distribution (full):")
    for cls in emotion_classes:
        n = (df["final_label"] == cls).sum()
        info_lines.append(f"  {cls:<12}: {n:>5}")

    with open(os.path.join(splits_dir, "split_info.txt"), "w") as f:
        f.write("\n".join(info_lines))

    print(f"\nAll saved to: {splits_dir}/")
    return train_df, val_df, test_df


# ============================================================
# ЗАГРУЗКА ГОТОВЫХ СПЛИТОВ
# ============================================================

def load_splits(
    splits_dir: str     = SPLITS_DIR,
    tokenizer_name: str = ROBERTA_NAME,
    max_len: int        = MAX_LEN,
    batch_size: int     = BATCH_SIZE,
):
    train_df = pd.read_csv(os.path.join(splits_dir, "train.csv"))
    val_df   = pd.read_csv(os.path.join(splits_dir, "val.csv"))
    test_df  = pd.read_csv(os.path.join(splits_dir, "test.csv"))
    map_df   = pd.read_csv(os.path.join(splits_dir, "label_mapping.csv"))

    # Проверяем маппинг
    for _, row in map_df.iterrows():
        expected = CLASS_TO_ID[row["class_name"]]
        actual   = int(row["class_id"])
        assert expected == actual, (
            f"Mapping mismatch: {row['class_name']} "
            f"expected id={expected}, got id={actual}. "
            f"Re-run prepare_and_save_splits()!"
        )

    num_labels = len(emotion_classes)
    samples_per_class = [int((train_df["label_id"] == i).sum()) for i in range(num_labels)]

    print(f"Splits loaded from: {splits_dir}/")
    print(f"  Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")
    print(f"\n  Train samples per class:")
    for i, cls in enumerate(emotion_classes):
        print(f"    id={i}  {cls:<12}: {samples_per_class[i]:>5}")

    tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)

    X_train, y_train = train_df["tweet_text"], train_df["label_id"]
    X_val,   y_val   = val_df["tweet_text"],   val_df["label_id"]
    X_test,  y_test  = test_df["tweet_text"],  test_df["label_id"]

    ds_train = TweetDataset(X_train, y_train, tokenizer, max_len)
    ds_val   = TweetDataset(X_val,   y_val,   tokenizer, max_len)
    ds_test  = TweetDataset(X_test,  y_test,  tokenizer, max_len)

    loaders = {
        "train": DataLoader(ds_train, batch_size=batch_size,     shuffle=True),
        "val":   DataLoader(ds_val,   batch_size=batch_size * 2, shuffle=False),
        "test":  DataLoader(ds_test,  batch_size=batch_size * 2, shuffle=False),
    }

    meta = {
        "num_labels":        num_labels,
        "samples_per_class": samples_per_class,
        "tokenizer":         tokenizer,
        "X_train": X_train, "y_train": y_train,
        "X_val":   X_val,   "y_val":   y_val,
        "X_test":  X_test,  "y_test":  y_test,
        "train_df": train_df,
        "val_df":   val_df,
        "test_df":  test_df,
    }

    return loaders, meta


# ============================================================
# ЗАПУСК ПОДГОТОВКИ
# ============================================================

train_df, val_df, test_df = prepare_and_save_splits(
    csv_path   = "dataset_final_verified_v2_neutral.csv",
    splits_dir = SPLITS_DIR,
)
```

```python
# ============================================================
# ЯЧЕЙКА 2: ЭКСПЕРИМЕНТЫ С ФУНКЦИЯМИ ПОТЕРЬ
# ============================================================

import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import re
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from torch.utils.data import DataLoader
from transformers import AutoModel, get_linear_schedule_with_warmup
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from copy import deepcopy

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME  = "roberta-base"
MAX_LEN       = 128
BATCH_SIZE    = 32
EPOCHS        = 10
LEARNING_RATE = 2e-5
WARMUP_RATIO  = 0.1
WEIGHT_DECAY  = 0.01
PATIENCE      = 3
SEED          = 42

emotion_classes = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]
CLASS_TO_ID     = {cls: i for i, cls in enumerate(emotion_classes)}
ID_TO_CLASS     = {i: cls for i, cls in enumerate(emotion_classes)}

OUTPUT_DIR = "loss_balancing_experiments"
os.makedirs(OUTPUT_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

# ============================================================
# ВОСПРОИЗВОДИМОСТЬ
# ============================================================

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

# ============================================================
# МОДЕЛЬ — RoBERTa
# ============================================================

class RobertaEmotionModel(nn.Module):
    def __init__(self, model_name, num_labels):
        super().__init__()
        self.backbone   = AutoModel.from_pretrained(model_name)
        self.dropout    = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.backbone.config.hidden_size, num_labels)

    def forward(self, input_ids, attention_mask):
        out = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0]
        return self.classifier(self.dropout(cls))

# ============================================================
# ФУНКЦИИ ПОТЕРЬ
# ============================================================

class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, weight=None, reduction="mean"):
        super().__init__()
        self.gamma     = gamma
        self.weight    = weight
        self.reduction = reduction

    def forward(self, logits, targets):
        log_prob = F.log_softmax(logits, dim=-1)
        prob     = log_prob.exp()
        log_pt   = log_prob.gather(1, targets.unsqueeze(1)).squeeze(1)
        pt       = prob.gather(1, targets.unsqueeze(1)).squeeze(1)
        loss     = -((1.0 - pt) ** self.gamma) * log_pt
        if self.weight is not None:
            loss = loss * self.weight.to(logits.device)[targets]
        return loss.mean() if self.reduction == "mean" else loss.sum()


class LabelSmoothingCrossEntropy(nn.Module):
    def __init__(self, smoothing=0.1, weight=None):
        super().__init__()
        self.smoothing = smoothing
        self.weight    = weight

    def forward(self, logits, targets):
        n_classes = logits.size(-1)
        log_prob  = F.log_softmax(logits, dim=-1)
        with torch.no_grad():
            dist = torch.full_like(log_prob, self.smoothing / (n_classes - 1))
            dist.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)
        loss = -(dist * log_prob).sum(dim=-1)
        if self.weight is not None:
            loss = loss * self.weight.to(logits.device)[targets]
        return loss.mean()


class ClassBalancedLoss(nn.Module):
    def __init__(self, samples_per_class, beta=None):
        super().__init__()
        if beta is None:
            N    = sum(samples_per_class)
            beta = (N - 1) / N
        eff     = [1.0 - beta ** n for n in samples_per_class]
        weights = [(1.0 - beta) / e for e in eff]
        total   = sum(weights)
        n_cls   = len(samples_per_class)
        weights = [w / total * n_cls for w in weights]
        self.register_buffer("weights", torch.tensor(weights, dtype=torch.float32))

    def forward(self, logits, targets):
        return F.cross_entropy(logits, targets, weight=self.weights.to(logits.device))


def build_weights_freq(spc, dev):
    f = np.array(spc, dtype=float)
    w = 1.0 / f
    w = w / w.sum() * len(spc)
    return torch.tensor(w, dtype=torch.float32).to(dev)


def build_weights_sqrt(spc, dev):
    f = np.array(spc, dtype=float)
    w = 1.0 / np.sqrt(f)
    w = w / w.sum() * len(spc)
    return torch.tensor(w, dtype=torch.float32).to(dev)

# ============================================================
# TRAIN / EVAL
# ============================================================

def train_epoch(model, loader, optimizer, scheduler, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []

    for batch in tqdm(loader, desc="  train", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        optimizer.zero_grad()
        logits = model(ids, mask)
        loss   = criterion(logits, labs)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()

        total_loss += loss.item()
        all_preds.extend(logits.argmax(1).cpu().numpy())
        all_labels.extend(labs.cpu().numpy())

    return total_loss / len(loader), f1_score(all_labels, all_preds, average="macro")


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = 0
    all_preds, all_labels, all_probs = [], [], []
    ce = nn.CrossEntropyLoss()

    for batch in tqdm(loader, desc="  eval", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)

        logits = model(ids, mask)
        total_loss += ce(logits, labs).item()
        probs = torch.softmax(logits, 1).cpu().numpy()

        all_preds.extend(probs.argmax(1))
        all_labels.extend(labs.cpu().numpy())
        all_probs.extend(probs)

    return (total_loss / len(loader),
            f1_score(all_labels, all_preds, average="macro"),
            np.array(all_preds), np.array(all_labels), np.array(all_probs))

# ============================================================
# УНИВЕРСАЛЬНЫЙ ЭКСПЕРИМЕНТ
# ============================================================

def run_experiment(train_df, val_df, test_df, display_name,
                   criterion_fn, tokenizer):
    print(f"\n{'='*60}")
    print(f"  {display_name}")
    print(f"{'='*60}")

    set_seed(SEED)
    num_labels = len(emotion_classes)

    X_train, y_train = train_df["tweet_text"], train_df["label_id"]
    X_val,   y_val   = val_df["tweet_text"],   val_df["label_id"]
    X_test,  y_test  = test_df["tweet_text"],  test_df["label_id"]

    samples_per_class = [int((y_train == i).sum()) for i in range(num_labels)]

    print(f"\n  Train size: {len(X_train)}")
    print(f"  Train class distribution:")
    for i, cls in enumerate(emotion_classes):
        print(f"    id={i}  {cls:<12}: {samples_per_class[i]:>5}")

    criterion = criterion_fn(samples_per_class)

    tr_loader = DataLoader(
        TweetDataset(X_train, y_train, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE, shuffle=True)
    va_loader = DataLoader(
        TweetDataset(X_val, y_val, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)
    te_loader = DataLoader(
        TweetDataset(X_test, y_test, tokenizer, MAX_LEN),
        batch_size=BATCH_SIZE * 2)

    set_seed(SEED)
    model     = RobertaEmotionModel(ROBERTA_NAME, num_labels).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = len(tr_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(
        optimizer, int(total_steps * WARMUP_RATIO), total_steps)

    best_f1, patience_cnt, best_state = 0.0, 0, None
    history = {"train_loss": [], "val_loss": [], "train_f1": [], "val_f1": []}

    for ep in range(1, EPOCHS + 1):
        tr_loss, tr_f1 = train_epoch(model, tr_loader, optimizer, scheduler, criterion)
        va_loss, va_f1, _, _, _ = evaluate(model, va_loader)

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(va_loss)
        history["train_f1"].append(tr_f1)
        history["val_f1"].append(va_f1)

        print(f"  Ep {ep}/{EPOCHS}  "
              f"train_loss={tr_loss:.4f} train_f1={tr_f1:.4f}  "
              f"val_loss={va_loss:.4f} val_f1={va_f1:.4f}", end="")

        if va_f1 > best_f1:
            best_f1, patience_cnt = va_f1, 0
            best_state = deepcopy(model.state_dict())
            print("  ✓ best")
        else:
            patience_cnt += 1
            print(f"  ⏳ {patience_cnt}/{PATIENCE}")
        if patience_cnt >= PATIENCE:
            print(f"  Early stop at epoch {ep}")
            break

    model.load_state_dict(best_state)
    te_loss, te_f1, te_preds, te_labels, te_probs = evaluate(model, te_loader)

    rep     = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4,
                                    output_dict=True)
    rep_str = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4)

    print(f"\n  Test loss={te_loss:.4f}  Macro F1={te_f1:.4f}")
    print(rep_str)

    result = {
        "variant":               display_name,
        "train_size":            len(X_train),
        "val_size":              len(X_val),
        "test_size":             len(X_test),
        "epochs_trained":        len(history["train_loss"]),
        "best_val_f1":           best_f1,
        "test_loss":             te_loss,
        "test_accuracy":         rep["accuracy"],
        "test_macro_precision":  rep["macro avg"]["precision"],
        "test_macro_recall":     rep["macro avg"]["recall"],
        "test_macro_f1":         rep["macro avg"]["f1-score"],
        "test_weighted_f1":      rep["weighted avg"]["f1-score"],
        "history":               history,
        "report_dict":           rep,
        "test_preds":            te_preds,
        "test_labels":           te_labels,
    }
    for cls in emotion_classes:
        result[f"f1_{cls}"] = rep[cls]["f1-score"] if cls in rep else 0.0

    return result

# ============================================================
# ЗАГРУЗКА ДАННЫХ
# ============================================================

loaders, meta = load_splits()

train_df = meta["train_df"]
val_df   = meta["val_df"]
test_df  = meta["test_df"]
tokenizer = meta["tokenizer"]

# ============================================================
# ОПРЕДЕЛЕНИЕ ЭКСПЕРИМЕНТОВ
# ============================================================

experiments = [
    {
        "display_name": "1. Baseline (no balancing)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(),
    },
    {
        "display_name": "2. Weighted CE (1/freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(weight=build_weights_freq(spc, device)),
    },
    {
        "display_name": "3. Weighted CE (1/sqrt freq)",
        "criterion_fn": lambda spc: nn.CrossEntropyLoss(weight=build_weights_sqrt(spc, device)),
    },
    {
        "display_name": "4. Focal Loss (gamma=2)",
        "criterion_fn": lambda spc: FocalLoss(gamma=2.0),
    },
    {
        "display_name": "5. Focal Loss + Weighted (1/freq)",
        "criterion_fn": lambda spc: FocalLoss(gamma=2.0, weight=build_weights_freq(spc, device)),
    },
    {
        "display_name": "6. Focal Loss + Weighted (1/sqrt freq)",
        "criterion_fn": lambda spc: FocalLoss(gamma=2.0, weight=build_weights_sqrt(spc, device)),
    },
    {
        "display_name": "7. Label Smoothing (eps=0.1)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(smoothing=0.1),
    },
    {
        "display_name": "8. Label Smoothing + Weighted (1/freq)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(
            smoothing=0.1, weight=build_weights_freq(spc, device)),
    },
    {
        "display_name": "9. Label Smoothing + Weighted (1/sqrt freq)",
        "criterion_fn": lambda spc: LabelSmoothingCrossEntropy(
            smoothing=0.1, weight=build_weights_sqrt(spc, device)),
    },
    {
        "display_name": "10. Class-Balanced Loss (beta auto)",
        "criterion_fn": lambda spc: ClassBalancedLoss(samples_per_class=spc),
    },
]


# ============================================================
# ЗАПУСК
# ============================================================

all_results = []
for exp in experiments:
    res = run_experiment(
        train_df, val_df, test_df,
        display_name = exp["display_name"],
        criterion_fn = exp["criterion_fn"],
        tokenizer    = tokenizer,
    )
    all_results.append(res)

# ============================================================
# ГРАФИКИ
# ============================================================

plt.rcParams['font.family'] = 'DejaVu Sans'
PROPS_BOX  = dict(boxstyle='round,pad=0.6', facecolor='#f5f5f5', edgecolor='#aaaaaa', alpha=0.9)
EXP_COLORS = ['#2c7bb6', '#1a9641', '#d7191c', '#fdae61',
              '#abd9e9', '#a6d96a', '#762a83', '#e66101']


def plot_training_curves(result, output_dir):
    ep_range = range(1, len(result["history"]["train_f1"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.patch.set_facecolor('white')

    for ax in axes:
        ax.set_facecolor('white')
        ax.grid(True, linestyle='--', alpha=0.4, zorder=1)
        ax.set_axisbelow(True)

    axes[0].plot(ep_range, result["history"]["train_f1"],
                 "o-", color="steelblue", lw=2.5, ms=6, label="Train Macro F1", zorder=3)
    axes[0].plot(ep_range, result["history"]["val_f1"],
                 "s-", color="#e74c3c", lw=2.5, ms=6, label="Val Macro F1", zorder=3)
    axes[0].set_title("Macro F1", fontsize=13, fontweight="bold", pad=12)
    axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Macro F1"); axes[0].legend(fontsize=11)

    axes[1].plot(ep_range, result["history"]["train_loss"],
                 "o-", color="steelblue", lw=2.5, ms=6, label="Train Loss", zorder=3)
    axes[1].plot(ep_range, result["history"]["val_loss"],
                 "s-", color="#e74c3c", lw=2.5, ms=6, label="Val Loss", zorder=3)
    axes[1].set_title("Loss", fontsize=13, fontweight="bold", pad=12)
    axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Loss"); axes[1].legend(fontsize=11)

    stats = (f"Epochs: {result['epochs_trained']}  |  "
             f"Best Val F1: {result['best_val_f1']:.4f}  |  "
             f"Train N: {result['train_size']:,}")
    axes[1].text(0.99, 0.03, stats, transform=axes[1].transAxes, fontsize=9,
                 va='bottom', ha='right', bbox=PROPS_BOX, family='monospace')

    plt.suptitle(result["variant"], fontsize=13, fontweight="bold", y=1.01)
    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"curves_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_per_class_f1(result, output_dir):
    rep     = result["report_dict"]
    f1_vals = [rep[cls]["f1-score"] if cls in rep else 0 for cls in emotion_classes]

    fig, ax = plt.subplots(figsize=(11, 6))
    fig.patch.set_facecolor('white'); ax.set_facecolor('white')

    bars = ax.bar(emotion_classes, f1_vals, color="steelblue", alpha=0.75, edgecolor='white', zorder=2)
    for b, v in zip(bars, f1_vals):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.012,
                f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

    ax.axhline(result["test_macro_f1"], color="#FFA500", ls="--", lw=2.2, zorder=3,
               label=f"Macro F1 = {result['test_macro_f1']:.4f}")
    ax.set_ylim(0, 1.05); ax.set_ylabel("F1-score"); ax.set_xlabel("Emotion Class")
    ax.set_title(f"Per-Class F1 — {result['variant']}", fontsize=12, fontweight="bold", pad=12)
    ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True)
    ax.legend(fontsize=11)

    stats = (f"Acc={result['test_accuracy']:.4f}  P={result['test_macro_precision']:.4f}  "
             f"R={result['test_macro_recall']:.4f}  F1={result['test_macro_f1']:.4f}  "
             f"wF1={result['test_weighted_f1']:.4f}")
    ax.text(0.99, 0.03, stats, transform=ax.transAxes, fontsize=9,
            va='bottom', ha='right', bbox=PROPS_BOX, family='monospace')

    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"per_class_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_confusion_matrix(result, output_dir):
    fig, ax = plt.subplots(figsize=(9, 7))
    fig.patch.set_facecolor('white')

    cm   = confusion_matrix(result["test_labels"], result["test_preds"])
    cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)

    sns.heatmap(cm_n, annot=True, fmt=".2f", cmap="Blues",
                xticklabels=emotion_classes, yticklabels=emotion_classes,
                ax=ax, linewidths=0.5, linecolor='white')
    ax.set_title(f"Confusion Matrix — {result['variant']}\n"
                 f"Macro F1={result['test_macro_f1']:.4f}  Acc={result['test_accuracy']:.4f}",
                 fontsize=11, fontweight="bold", pad=12)
    ax.set_ylabel("True"); ax.set_xlabel("Predicted")

    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"cm_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


for res in all_results:
    plot_training_curves(res, OUTPUT_DIR)
    plot_per_class_f1(res, OUTPUT_DIR)
    plot_confusion_matrix(res, OUTPUT_DIR)

# ---- Сравнительный Macro F1 ----
fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white'); ax.set_facecolor('white')

macro_f1s = [r["test_macro_f1"] for r in all_results]
bars = ax.bar(range(len(all_results)), macro_f1s, color=EXP_COLORS[:len(all_results)],
              alpha=0.82, edgecolor='white', zorder=2)
for b, v in zip(bars, macro_f1s):
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.003,
            f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")

ax.axhline(all_results[0]["test_macro_f1"], color='black', ls='--', lw=1.5,
           zorder=3, label=f"Baseline = {all_results[0]['test_macro_f1']:.4f}")
ax.set_xticks(range(len(all_results)))
ax.set_xticklabels([r["variant"] for r in all_results], rotation=30, ha='right', fontsize=10)
ax.set_ylabel("Macro F1"); ax.set_title("Loss Function Comparison — Macro F1 (RoBERTa)",
                                         fontsize=13, fontweight="bold", pad=12)
ax.set_ylim(max(0, min(macro_f1s)-0.05), min(1, max(macro_f1s)+0.06))
ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True); ax.legend(fontsize=11)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_macro_f1.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ---- Сравнительный Per-Class F1 ----
n_exp = len(all_results); x = np.arange(len(emotion_classes)); w = 0.10

fig, ax = plt.subplots(figsize=(16, 7))
fig.patch.set_facecolor('white'); ax.set_facecolor('white')

for i, (res, color) in enumerate(zip(all_results, EXP_COLORS)):
    f1s    = [res.get(f"f1_{c}", 0) for c in emotion_classes]
    offset = (i - n_exp/2 + 0.5) * w
    ax.bar(x + offset, f1s, w, label=res["variant"], color=color, alpha=0.80, edgecolor='white', zorder=2)

ax.set_ylabel("F1-score"); ax.set_xlabel("Emotion Class")
ax.set_title("Per-Class F1 — All Loss Methods", fontsize=13, fontweight="bold", pad=12)
ax.set_xticks(x); ax.set_xticklabels(emotion_classes, fontsize=12)
ax.set_ylim(0, 1.10)
ax.legend(fontsize=8, ncol=2, loc='upper right')
ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_per_class_f1_all.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ---- Тепловая карта Delta ----
baseline_pc = {cls: all_results[0].get(f"f1_{cls}", 0) for cls in emotion_classes}
delta_rows  = []
for res in all_results[1:]:
    row = {"Method": res["variant"]}
    for cls in emotion_classes:
        row[cls] = round(res.get(f"f1_{cls}", 0) - baseline_pc[cls], 4)
    delta_rows.append(row)

delta_df = pd.DataFrame(delta_rows).set_index("Method")

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('white')
sns.heatmap(delta_df, annot=True, fmt=".4f", cmap="RdYlGn", center=0,
            linewidths=0.5, linecolor='white', ax=ax, cbar_kws={"label": "ΔF1 vs Baseline"})
ax.set_title("Per-Class F1 Delta vs Baseline\nGreen=improvement, Red=degradation",
             fontsize=12, fontweight="bold", pad=12)
ax.set_xlabel("Emotion Class"); ax.set_ylabel("Loss Method")

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "delta_heatmap.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ИТОГОВЫЕ ТАБЛИЦЫ
# ============================================================

print(f"\n{'='*100}")
print("  LOSS FUNCTION COMPARISON — RoBERTa (FIXED LABEL MAPPING)")
print(f"{'='*100}\n")

rows = []
for r in all_results:
    rows.append({
        "Method":       r["variant"],
        "Train N":      r["train_size"],
        "Epochs":       r["epochs_trained"],
        "Accuracy":     round(r["test_accuracy"], 4),
        "Macro P":      round(r["test_macro_precision"], 4),
        "Macro R":      round(r["test_macro_recall"], 4),
        "Macro F1":     round(r["test_macro_f1"], 4),
        "Weighted F1":  round(r["test_weighted_f1"], 4),
        "Best Val F1":  round(r["best_val_f1"], 4),
        "Test Loss":    round(r["test_loss"], 4),
    })

summary_df = pd.DataFrame(rows).set_index("Method")
print(summary_df.to_string())

print(f"\n{'='*100}")
print("  DELTA vs BASELINE")
print(f"{'='*100}\n")

bl = all_results[0]["test_macro_f1"]
for r in all_results[1:]:
    d = r["test_macro_f1"] - bl
    print(f"  {r['variant']:<50}  Δ = {'+' if d>=0 else ''}{d:.4f}")

print(f"\n{'='*100}")
print("  PER-CLASS F1")
print(f"{'='*100}\n")

pc_rows = []
for r in all_results:
    row = {"Method": r["variant"]}
    for cls in emotion_classes:
        row[cls] = round(r.get(f"f1_{cls}", 0), 4)
    pc_rows.append(row)

pc_df = pd.DataFrame(pc_rows).set_index("Method")
print(pc_df.to_string())

print(f"\n{'='*100}")
print("  PER-CLASS DELTA")
print(f"{'='*100}\n")
print(delta_df.to_string())

# Сохранение
summary_df.to_csv(os.path.join(OUTPUT_DIR, "summary_metrics.csv"))
pc_df.to_csv(os.path.join(OUTPUT_DIR, "per_class_f1.csv"))
delta_df.to_csv(os.path.join(OUTPUT_DIR, "delta_per_class_f1.csv"))

best_idx = int(np.argmax([r["test_macro_f1"] for r in all_results]))
print(f"\n🏆 Best: {all_results[best_idx]['variant']}")
print(f"   Macro F1 = {all_results[best_idx]['test_macro_f1']:.4f}")
print(f"\nSaved to: {OUTPUT_DIR}/")

# В конце ячейки 2 — сохраняем лучший лосс
best_idx = int(np.argmax([r["test_macro_f1"] for r in all_results]))
BEST_LOSS_NAME = experiments[best_idx]["display_name"]
BEST_LOSS_FN   = experiments[best_idx]["criterion_fn"]

print(f"\n🏆 Best loss: {BEST_LOSS_NAME}")
print(f"   Will be used in augmentation experiments")
```

```python
# ============================================================
# ЯЧЕЙКА 3: ЭКСПЕРИМЕНТЫ С АУГМЕНТАЦИЕЙ ДАННЫХ
# ============================================================

import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import re
import random
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from torch.utils.data import DataLoader
from transformers import (
    AutoModel, AutoTokenizer,
    MarianMTModel, MarianTokenizer,
    get_linear_schedule_with_warmup,
)
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from copy import deepcopy

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME  = "roberta-base"
MAX_LEN       = 128
BATCH_SIZE    = 32
EPOCHS        = 10
LEARNING_RATE = 2e-5
WARMUP_RATIO  = 0.1
WEIGHT_DECAY  = 0.01
PATIENCE      = 3
SEED          = 42

emotion_classes  = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]
CLASS_TO_ID      = {cls: i for i, cls in enumerate(emotion_classes)}
ID_TO_CLASS      = {i: cls for i, cls in enumerate(emotion_classes)}
MINORITY_CLASSES = ["fear", "disgust", "surprise"]
TARGET_N         = 1000

SPLITS_DIR = "splits"
AUG_DIR    = "augmented_data"
OUTPUT_DIR = "data_augmentation_experiments"
os.makedirs(OUTPUT_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

def set_seed(seed):
    np.random.seed(seed)
    random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

# ============================================================
# МОДЕЛЬ + ЛОСС (LS + WCE 1/freq)
# ============================================================

class RobertaEmotionModel(nn.Module):
    def __init__(self, model_name, num_labels):
        super().__init__()
        self.backbone   = AutoModel.from_pretrained(model_name)
        self.dropout    = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.backbone.config.hidden_size, num_labels)

    def forward(self, input_ids, attention_mask):
        out = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0]
        return self.classifier(self.dropout(cls))


class LabelSmoothingCrossEntropy(nn.Module):
    def __init__(self, smoothing=0.1, weight=None):
        super().__init__()
        self.smoothing = smoothing
        self.weight    = weight

    def forward(self, logits, targets):
        n_classes = logits.size(-1)
        log_prob  = F.log_softmax(logits, dim=-1)
        with torch.no_grad():
            dist = torch.full_like(log_prob, self.smoothing / (n_classes - 1))
            dist.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)
        loss = -(dist * log_prob).sum(dim=-1)
        if self.weight is not None:
            loss = loss * self.weight.to(logits.device)[targets]
        return loss.mean()


def build_weights_freq(spc, dev):
    f = np.array(spc, dtype=float)
    w = 1.0 / f
    w = w / w.sum() * len(spc)
    return torch.tensor(w, dtype=torch.float32).to(dev)


def build_criterion(spc):
    return LabelSmoothingCrossEntropy(smoothing=0.1)

# ============================================================
# TRAIN / EVAL
# ============================================================

def train_epoch(model, loader, optimizer, scheduler, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []
    for batch in tqdm(loader, desc="  train", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)
        optimizer.zero_grad()
        logits = model(ids, mask)
        loss   = criterion(logits, labs)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        all_preds.extend(logits.argmax(1).cpu().numpy())
        all_labels.extend(labs.cpu().numpy())
    return total_loss / len(loader), f1_score(all_labels, all_preds, average="macro")


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = 0
    all_preds, all_labels, all_probs = [], [], []
    ce = nn.CrossEntropyLoss()
    for batch in tqdm(loader, desc="  eval", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)
        logits = model(ids, mask)
        total_loss += ce(logits, labs).item()
        probs = torch.softmax(logits, 1).cpu().numpy()
        all_preds.extend(probs.argmax(1))
        all_labels.extend(labs.cpu().numpy())
        all_probs.extend(probs)
    return (total_loss / len(loader),
            f1_score(all_labels, all_preds, average="macro"),
            np.array(all_preds), np.array(all_labels), np.array(all_probs))

# ============================================================
# УНИВЕРСАЛЬНЫЙ ЭКСПЕРИМЕНТ
# ============================================================

def run_experiment(train_df, val_df, test_df, display_name, tokenizer):
    print(f"\n{'='*60}")
    print(f"  {display_name}")
    print(f"{'='*60}")

    set_seed(SEED)
    num_labels = len(emotion_classes)

    X_train, y_train = train_df["tweet_text"], train_df["label_id"]
    X_val,   y_val   = val_df["tweet_text"],   val_df["label_id"]
    X_test,  y_test  = test_df["tweet_text"],  test_df["label_id"]

    spc = [int((y_train == i).sum()) for i in range(num_labels)]

    print(f"\n  Train size: {len(X_train)}")
    for i, cls in enumerate(emotion_classes):
        print(f"    id={i}  {cls:<12}: {spc[i]:>5}")

    criterion = build_criterion(spc)

    tr_loader = DataLoader(TweetDataset(X_train, y_train, tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE, shuffle=True)
    va_loader = DataLoader(TweetDataset(X_val, y_val, tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE*2)
    te_loader = DataLoader(TweetDataset(X_test, y_test, tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE*2)

    set_seed(SEED)
    model     = RobertaEmotionModel(ROBERTA_NAME, num_labels).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = len(tr_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(optimizer, int(total_steps*WARMUP_RATIO), total_steps)

    best_f1, patience_cnt, best_state = 0.0, 0, None
    history = {"train_loss": [], "val_loss": [], "train_f1": [], "val_f1": []}

    for ep in range(1, EPOCHS+1):
        tr_loss, tr_f1 = train_epoch(model, tr_loader, optimizer, scheduler, criterion)
        va_loss, va_f1, _, _, _ = evaluate(model, va_loader)
        history["train_loss"].append(tr_loss); history["val_loss"].append(va_loss)
        history["train_f1"].append(tr_f1);     history["val_f1"].append(va_f1)

        print(f"  Ep {ep}/{EPOCHS}  "
              f"trL={tr_loss:.4f} trF1={tr_f1:.4f}  "
              f"vaL={va_loss:.4f} vaF1={va_f1:.4f}", end="")

        if va_f1 > best_f1:
            best_f1, patience_cnt = va_f1, 0
            best_state = deepcopy(model.state_dict())
            print("  ✓")
        else:
            patience_cnt += 1
            print(f"  ⏳ {patience_cnt}/{PATIENCE}")
        if patience_cnt >= PATIENCE:
            print(f"  Early stop at epoch {ep}"); break

    model.load_state_dict(best_state)
    te_loss, te_f1, te_preds, te_labels, _ = evaluate(model, te_loader)

    rep     = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4, output_dict=True)
    rep_str = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4)
    print(f"\n  Test loss={te_loss:.4f}  Macro F1={te_f1:.4f}")
    print(rep_str)

    result = {
        "variant": display_name, "train_size": len(X_train),
        "val_size": len(X_val), "test_size": len(X_test),
        "epochs_trained": len(history["train_loss"]), "best_val_f1": best_f1,
        "test_loss": te_loss, "test_accuracy": rep["accuracy"],
        "test_macro_precision": rep["macro avg"]["precision"],
        "test_macro_recall":    rep["macro avg"]["recall"],
        "test_macro_f1":        rep["macro avg"]["f1-score"],
        "test_weighted_f1":     rep["weighted avg"]["f1-score"],
        "history": history, "report_dict": rep,
        "test_preds": te_preds, "test_labels": te_labels,
    }
    for cls in emotion_classes:
        result[f"f1_{cls}"] = rep[cls]["f1-score"] if cls in rep else 0.0
    return result

# ============================================================
# МЕТОДЫ АУГМЕНТАЦИИ
# ============================================================

def random_oversampling(train_df, target_n=TARGET_N, minority=MINORITY_CLASSES):
    print(f"\n🔄 Random Oversampling → target={target_n}")
    aug = train_df.copy()
    for cls in minority:
        cls_df = train_df[train_df["final_label"] == cls]
        need   = target_n - len(cls_df)
        if need <= 0:
            print(f"  {cls}: already {len(cls_df)} >= {target_n}, skip"); continue
        extra = cls_df.sample(n=need, replace=True, random_state=SEED)
        aug   = pd.concat([aug, extra], ignore_index=True)
        print(f"  {cls}: {len(cls_df)} → {len(cls_df)+need} (+{need})")
    aug = aug.sample(frac=1, random_state=SEED).reset_index(drop=True)
    print(f"  Total: {len(train_df)} → {len(aug)}")
    return aug


@torch.no_grad()
def back_translate_batch(texts, tok_fwd, model_fwd, tok_bwd, model_bwd, max_len=128):
    inp = tok_fwd(texts, return_tensors="pt", padding=True, truncation=True, max_length=max_len).to(device)
    mid = model_fwd.generate(**inp, max_length=max_len, num_beams=4, do_sample=True, temperature=0.9, top_p=0.95)
    mid_texts = tok_fwd.batch_decode(mid, skip_special_tokens=True)
    inp2 = tok_bwd(mid_texts, return_tensors="pt", padding=True, truncation=True, max_length=max_len).to(device)
    out  = model_bwd.generate(**inp2, max_length=max_len, num_beams=4, do_sample=True, temperature=0.9, top_p=0.95)
    return tok_bwd.batch_decode(out, skip_special_tokens=True)


def back_translation_augmentation(train_df, target_n=TARGET_N,
                                   minority=MINORITY_CLASSES, bt_bs=32):
    print(f"\n🔄 Back-Translation → target={target_n}")
    print("  Loading translation models (en↔de)...")
    tok_fwd   = MarianTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-de")
    model_fwd = MarianMTModel.from_pretrained("Helsinki-NLP/opus-mt-en-de").to(device).eval()
    tok_bwd   = MarianTokenizer.from_pretrained("Helsinki-NLP/opus-mt-de-en")
    model_bwd = MarianMTModel.from_pretrained("Helsinki-NLP/opus-mt-de-en").to(device).eval()
    print("  ✅ Models loaded")

    aug = train_df.copy()

    for cls in minority:
        cls_df = train_df[train_df["final_label"] == cls]
        need   = target_n - len(cls_df)
        if need <= 0:
            print(f"  {cls}: skip"); continue

        print(f"\n  🔄 {cls}: generating {need} via BT")
        sources = cls_df["tweet_text"].tolist()
        selected = random.choices(sources, k=need) if need > len(sources) else random.sample(sources, k=need)
        label_id = cls_df["label_id"].iloc[0]

        bt_results = []
        for i in tqdm(range(0, len(selected), bt_bs), desc=f"    BT {cls}", leave=False):
            batch = selected[i:i+bt_bs]
            bt_results.extend(back_translate_batch(batch, tok_fwd, model_fwd, tok_bwd, model_bwd))

        valid = []
        for orig, bt in zip(selected, bt_results):
            bt = bt.strip()
            if len(bt) < 10 or bt.lower() == orig.lower():
                continue
            valid.append({"tweet_text": bt, "label_id": label_id, "final_label": cls})

        # Добиваем если не хватило
        if len(valid) < need:
            short = need - len(valid)
            extra = cls_df.sample(n=short, replace=True, random_state=SEED)
            valid.extend(extra[["tweet_text", "label_id", "final_label"]].to_dict("records"))
            print(f"    ⚠️ BT valid={len(valid)-short}, padded +{short}")

        aug = pd.concat([aug, pd.DataFrame(valid[:need])], ignore_index=True)
        print(f"    ✅ {cls}: +{min(len(valid), need)}")

    del tok_fwd, model_fwd, tok_bwd, model_bwd
    torch.cuda.empty_cache()

    aug = aug.sample(frac=1, random_state=SEED).reset_index(drop=True)
    print(f"\n  Total: {len(train_df)} → {len(aug)}")
    return aug


def load_llm_augmented():
    path = os.path.join(AUG_DIR, "train_augmented.csv")
    df   = pd.read_csv(path)

    # Проверяем/исправляем label_id по фиксированному маппингу
    df["label_id"] = df["final_label"].map(CLASS_TO_ID)
    assert df["label_id"].isna().sum() == 0, "Unmapped labels in augmented data!"

    print(f"\n📥 LLM-augmented train: {len(df)} from {path}")
    for cls in emotion_classes:
        print(f"  id={CLASS_TO_ID[cls]}  {cls:<12}: {(df['final_label']==cls).sum():>5}")
    return df

# ============================================================
# ЗАГРУЗКА БАЗОВЫХ СПЛИТОВ
# ============================================================

loaders, meta = load_splits()
train_df  = meta["train_df"]
val_df    = meta["val_df"]
test_df   = meta["test_df"]
tokenizer = meta["tokenizer"]

# ============================================================
# ЗАПУСК 4 ЭКСПЕРИМЕНТОВ
# ============================================================

all_results = []

# 1. Baseline
res1 = run_experiment(train_df, val_df, test_df,
                      "1. Baseline (Label Smothing)", tokenizer)
all_results.append(res1)

# 2. Random Oversampling
train_ros = random_oversampling(train_df)
res2 = run_experiment(train_ros, val_df, test_df,
                      "2. Random Oversampling", tokenizer)
all_results.append(res2)
del train_ros

# 3. Back-Translation
train_bt = back_translation_augmentation(train_df)
res3 = run_experiment(train_bt, val_df, test_df,
                      "3. Back-Translation (en-de-en)", tokenizer)
all_results.append(res3)
del train_bt

# 4. LLM Augmentation
train_llm = load_llm_augmented()
res4 = run_experiment(train_llm, val_df, test_df,
                      "4. LLM Augmentation (DeepSeek)", tokenizer)
all_results.append(res4)
del train_llm

# ============================================================
# ГРАФИКИ
# ============================================================

plt.rcParams['font.family'] = 'DejaVu Sans'
PROPS_BOX  = dict(boxstyle='round,pad=0.6', facecolor='#f5f5f5', edgecolor='#aaaaaa', alpha=0.9)
EXP_COLORS = ['#2c7bb6', '#1a9641', '#d7191c', '#fdae61']


def plot_training_curves(result, output_dir):
    ep = range(1, len(result["history"]["train_f1"])+1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6)); fig.patch.set_facecolor('white')
    for ax in axes:
        ax.set_facecolor('white'); ax.grid(True, ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True)
    axes[0].plot(ep, result["history"]["train_f1"], "o-", color="steelblue", lw=2.5, ms=6, label="Train F1", zorder=3)
    axes[0].plot(ep, result["history"]["val_f1"],   "s-", color="#e74c3c",   lw=2.5, ms=6, label="Val F1",   zorder=3)
    axes[0].set_title("Macro F1", fontsize=13, fontweight="bold"); axes[0].legend(fontsize=11)
    axes[1].plot(ep, result["history"]["train_loss"], "o-", color="steelblue", lw=2.5, ms=6, label="Train Loss", zorder=3)
    axes[1].plot(ep, result["history"]["val_loss"],   "s-", color="#e74c3c",   lw=2.5, ms=6, label="Val Loss",   zorder=3)
    axes[1].set_title("Loss", fontsize=13, fontweight="bold"); axes[1].legend(fontsize=11)
    plt.suptitle(result["variant"], fontsize=13, fontweight="bold", y=1.01); plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"curves_{safe}.png"), dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_per_class_f1(result, output_dir):
    rep = result["report_dict"]
    f1s = [rep[c]["f1-score"] if c in rep else 0 for c in emotion_classes]
    fig, ax = plt.subplots(figsize=(11, 6)); fig.patch.set_facecolor('white'); ax.set_facecolor('white')
    bars = ax.bar(emotion_classes, f1s, color="steelblue", alpha=0.75, edgecolor='white', zorder=2)
    for b, v in zip(bars, f1s):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.012, f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")
    ax.axhline(result["test_macro_f1"], color="#FFA500", ls="--", lw=2.2, zorder=3,
               label=f"Macro F1={result['test_macro_f1']:.4f}")
    ax.set_ylim(0, 1.05); ax.set_ylabel("F1"); ax.set_xlabel("Class")
    ax.set_title(f"Per-Class F1 — {result['variant']}", fontsize=12, fontweight="bold", pad=12)
    ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True); ax.legend(fontsize=11)
    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"perclass_{safe}.png"), dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_confusion_matrix(result, output_dir):
    fig, ax = plt.subplots(figsize=(9, 7)); fig.patch.set_facecolor('white')
    cm = confusion_matrix(result["test_labels"], result["test_preds"])
    cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)
    sns.heatmap(cm_n, annot=True, fmt=".2f", cmap="Blues",
                xticklabels=emotion_classes, yticklabels=emotion_classes, ax=ax, linewidths=0.5, linecolor='white')
    ax.set_title(f"CM — {result['variant']}\nF1={result['test_macro_f1']:.4f} Acc={result['test_accuracy']:.4f}",
                 fontsize=11, fontweight="bold", pad=12)
    ax.set_ylabel("True"); ax.set_xlabel("Predicted"); plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"cm_{safe}.png"), dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


for res in all_results:
    plot_training_curves(res, OUTPUT_DIR)
    plot_per_class_f1(res, OUTPUT_DIR)
    plot_confusion_matrix(res, OUTPUT_DIR)

# ---- Сравнительный Macro F1 ----
fig, ax = plt.subplots(figsize=(12, 6)); fig.patch.set_facecolor('white'); ax.set_facecolor('white')
mf1s = [r["test_macro_f1"] for r in all_results]
bars = ax.bar(range(len(all_results)), mf1s, color=EXP_COLORS, alpha=0.82, edgecolor='white', zorder=2)
for b, v in zip(bars, mf1s):
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.003, f"{v:.4f}", ha="center", fontsize=11, fontweight="bold")
ax.axhline(mf1s[0], color='black', ls='--', lw=1.5, zorder=3, label=f"Baseline={mf1s[0]:.4f}")
ax.set_xticks(range(len(all_results)))
ax.set_xticklabels([r["variant"] for r in all_results], rotation=20, ha='right', fontsize=10)
ax.set_ylabel("Macro F1"); ax.set_title("Augmentation Comparison — Macro F1", fontsize=13, fontweight="bold", pad=12)
ax.set_ylim(max(0, min(mf1s)-0.05), min(1, max(mf1s)+0.06))
ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True); ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_macro_f1.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ---- Per-Class F1 grouped ----
n_exp = len(all_results); x = np.arange(len(emotion_classes)); w = 0.18
fig, ax = plt.subplots(figsize=(15, 7)); fig.patch.set_facecolor('white'); ax.set_facecolor('white')
for i, (res, col) in enumerate(zip(all_results, EXP_COLORS)):
    f1s = [res.get(f"f1_{c}", 0) for c in emotion_classes]
    off = (i - n_exp/2 + 0.5) * w
    bars = ax.bar(x+off, f1s, w, label=res["variant"], color=col, alpha=0.80, edgecolor='white', zorder=2)
    for b, v in zip(bars, f1s):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.005, f"{v:.2f}", ha="center", fontsize=7, rotation=45)
ax.set_xticks(x); ax.set_xticklabels(emotion_classes, fontsize=12)
ax.set_ylim(0, 1.10); ax.set_ylabel("F1"); ax.set_xlabel("Class")
ax.set_title("Per-Class F1 — Augmentation Methods", fontsize=13, fontweight="bold", pad=12)
ax.legend(fontsize=9, loc='upper right'); ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "comparison_per_class_f1.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ---- Delta heatmap ----
bl_pc = {c: all_results[0].get(f"f1_{c}", 0) for c in emotion_classes}
delta_rows = [{"Method": r["variant"],
               **{c: round(r.get(f"f1_{c}", 0)-bl_pc[c], 4) for c in emotion_classes}}
              for r in all_results[1:]]
delta_df = pd.DataFrame(delta_rows).set_index("Method")

fig, ax = plt.subplots(figsize=(14, 5)); fig.patch.set_facecolor('white')
sns.heatmap(delta_df, annot=True, fmt=".4f", cmap="RdYlGn", center=0,
            linewidths=0.5, linecolor='white', ax=ax, cbar_kws={"label": "ΔF1"})
ax.set_title("Per-Class F1 Delta vs Baseline\nGreen=better, Red=worse",
             fontsize=12, fontweight="bold", pad=12)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "delta_heatmap.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ---- Minority focus ----
fig, ax = plt.subplots(figsize=(12, 6)); fig.patch.set_facecolor('white'); ax.set_facecolor('white')
xm = np.arange(len(MINORITY_CLASSES)); wm = 0.18
for i, (res, col) in enumerate(zip(all_results, EXP_COLORS)):
    f1s = [res.get(f"f1_{c}", 0) for c in MINORITY_CLASSES]
    off = (i - n_exp/2 + 0.5) * wm
    bars = ax.bar(xm+off, f1s, wm, label=res["variant"], color=col, alpha=0.80, edgecolor='white', zorder=2)
    for b, v in zip(bars, f1s):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.008, f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")
ax.set_xticks(xm); ax.set_xticklabels(MINORITY_CLASSES, fontsize=13)
ax.set_ylim(0, 1.10); ax.set_ylabel("F1")
ax.set_title("Minority Classes F1 — Augmentation Impact", fontsize=13, fontweight="bold", pad=12)
ax.legend(fontsize=9); ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "minority_comparison.png"), dpi=150, bbox_inches="tight", facecolor='white')
plt.show()

# ============================================================
# ТАБЛИЦЫ
# ============================================================

print(f"\n{'='*100}")
print("  DATA AUGMENTATION — RoBERTa + LS+WCE (1/freq) — FIXED MAPPING")
print(f"{'='*100}\n")

rows = [{"Method": r["variant"], "Train N": r["train_size"], "Epochs": r["epochs_trained"],
         "Acc": round(r["test_accuracy"],4), "MacroP": round(r["test_macro_precision"],4),
         "MacroR": round(r["test_macro_recall"],4), "MacroF1": round(r["test_macro_f1"],4),
         "wF1": round(r["test_weighted_f1"],4), "ValF1": round(r["best_val_f1"],4)}
        for r in all_results]
summary_df = pd.DataFrame(rows).set_index("Method")
print(summary_df.to_string())

print(f"\n  DELTA vs BASELINE:")
bl = all_results[0]["test_macro_f1"]
for r in all_results[1:]:
    d = r["test_macro_f1"] - bl
    print(f"    {r['variant']:<50}  Δ = {'+' if d>=0 else ''}{d:.4f}")

print(f"\n  PER-CLASS F1:")
pc_rows = [{"Method": r["variant"], **{c: round(r.get(f"f1_{c}",0),4) for c in emotion_classes}}
           for r in all_results]
pc_df = pd.DataFrame(pc_rows).set_index("Method")
print(pc_df.to_string())

print(f"\n  PER-CLASS DELTA:")
print(delta_df.to_string())

summary_df.to_csv(os.path.join(OUTPUT_DIR, "summary_metrics.csv"))
pc_df.to_csv(os.path.join(OUTPUT_DIR, "per_class_f1.csv"))
delta_df.to_csv(os.path.join(OUTPUT_DIR, "delta_per_class_f1.csv"))

best_idx = int(np.argmax([r["test_macro_f1"] for r in all_results]))
print(f"\n🏆 Best: {all_results[best_idx]['variant']}")
print(f"   Macro F1 = {all_results[best_idx]['test_macro_f1']:.4f}")
print(f"\nSaved to: {OUTPUT_DIR}/")
```

```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import re
import random
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from torch.utils.data import DataLoader
from transformers import AutoModel, AutoTokenizer, get_linear_schedule_with_warmup
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from copy import deepcopy

# ============================================================
# КОНФИГУРАЦИЯ
# ============================================================

ROBERTA_NAME  = "roberta-base"
MAX_LEN       = 128
BATCH_SIZE    = 32
EPOCHS        = 10
LEARNING_RATE = 2e-5
WARMUP_RATIO  = 0.1
WEIGHT_DECAY  = 0.01
PATIENCE      = 3
SEED          = 42

emotion_classes  = ["neutral", "happiness", "anger", "disgust", "sadness", "surprise", "fear"]
CLASS_TO_ID      = {cls: i for i, cls in enumerate(emotion_classes)}
MINORITY_CLASSES = ["fear", "disgust", "surprise"]

SPLITS_DIR = "splits"
AUG_DIR    = "augmented_data"
OUTPUT_DIR = "data_augmentation_experiments"
os.makedirs(OUTPUT_DIR, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

def set_seed(seed):
    np.random.seed(seed)
    random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True

# ============================================================
# МОДЕЛЬ
# ============================================================

class RobertaEmotionModel(nn.Module):
    def __init__(self, model_name, num_labels):
        super().__init__()
        self.backbone   = AutoModel.from_pretrained(model_name)
        self.dropout    = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.backbone.config.hidden_size, num_labels)

    def forward(self, input_ids, attention_mask):
        out = self.backbone(input_ids=input_ids, attention_mask=attention_mask)
        cls = out.last_hidden_state[:, 0]
        return self.classifier(self.dropout(cls))

# ============================================================
# ЛОСС
# ============================================================

class LabelSmoothingCrossEntropy(nn.Module):
    def __init__(self, smoothing=0.1, weight=None):
        super().__init__()
        self.smoothing = smoothing
        self.weight    = weight

    def forward(self, logits, targets):
        n_classes = logits.size(-1)
        log_prob  = F.log_softmax(logits, dim=-1)
        with torch.no_grad():
            dist = torch.full_like(log_prob, self.smoothing / (n_classes - 1))
            dist.scatter_(1, targets.unsqueeze(1), 1.0 - self.smoothing)
        loss = -(dist * log_prob).sum(dim=-1)
        if self.weight is not None:
            loss = loss * self.weight.to(logits.device)[targets]
        return loss.mean()

def build_criterion(spc):
    return LabelSmoothingCrossEntropy(smoothing=0.1)

# ============================================================
# TRAIN / EVAL
# ============================================================

def train_epoch(model, loader, optimizer, scheduler, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []
    for batch in tqdm(loader, desc="  train", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)
        optimizer.zero_grad()
        logits = model(ids, mask)
        loss   = criterion(logits, labs)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
        all_preds.extend(logits.argmax(1).cpu().numpy())
        all_labels.extend(labs.cpu().numpy())
    return total_loss / len(loader), f1_score(all_labels, all_preds, average="macro")


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = 0
    all_preds, all_labels, all_probs = [], [], []
    ce = nn.CrossEntropyLoss()
    for batch in tqdm(loader, desc="  eval", leave=False):
        ids  = batch["input_ids"].to(device)
        mask = batch["attention_mask"].to(device)
        labs = batch["labels"].to(device)
        logits = model(ids, mask)
        total_loss += ce(logits, labs).item()
        probs = torch.softmax(logits, 1).cpu().numpy()
        all_preds.extend(probs.argmax(1))
        all_labels.extend(labs.cpu().numpy())
        all_probs.extend(probs)
    return (total_loss / len(loader),
            f1_score(all_labels, all_preds, average="macro"),
            np.array(all_preds), np.array(all_labels), np.array(all_probs))

# ============================================================
# ЭКСПЕРИМЕНТ
# ============================================================

def run_experiment(train_df, val_df, test_df, display_name, tokenizer):
    print(f"\n{'='*60}")
    print(f"  {display_name}")
    print(f"{'='*60}")

    set_seed(SEED)
    num_labels = len(emotion_classes)

    X_train, y_train = train_df["tweet_text"], train_df["label_id"]
    X_val,   y_val   = val_df["tweet_text"],   val_df["label_id"]
    X_test,  y_test  = test_df["tweet_text"],  test_df["label_id"]

    spc = [int((y_train == i).sum()) for i in range(num_labels)]

    print(f"\n  Train size: {len(X_train)}")
    for i, cls in enumerate(emotion_classes):
        print(f"    id={i}  {cls:<12}: {spc[i]:>5}")

    criterion = build_criterion(spc)

    tr_loader = DataLoader(TweetDataset(X_train, y_train, tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE, shuffle=True)
    va_loader = DataLoader(TweetDataset(X_val,   y_val,   tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE * 2)
    te_loader = DataLoader(TweetDataset(X_test,  y_test,  tokenizer, MAX_LEN),
                           batch_size=BATCH_SIZE * 2)

    set_seed(SEED)
    model     = RobertaEmotionModel(ROBERTA_NAME, num_labels).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    total_steps = len(tr_loader) * EPOCHS
    scheduler   = get_linear_schedule_with_warmup(
        optimizer, int(total_steps * WARMUP_RATIO), total_steps)

    best_f1, patience_cnt, best_state = 0.0, 0, None
    history = {"train_loss": [], "val_loss": [], "train_f1": [], "val_f1": []}

    for ep in range(1, EPOCHS + 1):
        tr_loss, tr_f1 = train_epoch(model, tr_loader, optimizer, scheduler, criterion)
        va_loss, va_f1, _, _, _ = evaluate(model, va_loader)
        history["train_loss"].append(tr_loss); history["val_loss"].append(va_loss)
        history["train_f1"].append(tr_f1);     history["val_f1"].append(va_f1)

        print(f"  Ep {ep}/{EPOCHS}  "
              f"trL={tr_loss:.4f} trF1={tr_f1:.4f}  "
              f"vaL={va_loss:.4f} vaF1={va_f1:.4f}", end="")

        if va_f1 > best_f1:
            best_f1, patience_cnt = va_f1, 0
            best_state = deepcopy(model.state_dict())
            print("  ✓")
        else:
            patience_cnt += 1
            print(f"  ⏳ {patience_cnt}/{PATIENCE}")
        if patience_cnt >= PATIENCE:
            print(f"  Early stop at epoch {ep}"); break

    model.load_state_dict(best_state)
    te_loss, te_f1, te_preds, te_labels, _ = evaluate(model, te_loader)

    rep     = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4, output_dict=True)
    rep_str = classification_report(te_labels, te_preds,
                                    target_names=emotion_classes, digits=4)
    print(f"\n  Test loss={te_loss:.4f}  Macro F1={te_f1:.4f}")
    print(rep_str)

    result = {
        "variant": display_name, "train_size": len(X_train),
        "val_size": len(X_val),   "test_size":  len(X_test),
        "epochs_trained": len(history["train_loss"]), "best_val_f1": best_f1,
        "test_loss": te_loss, "test_accuracy": rep["accuracy"],
        "test_macro_precision": rep["macro avg"]["precision"],
        "test_macro_recall":    rep["macro avg"]["recall"],
        "test_macro_f1":        rep["macro avg"]["f1-score"],
        "test_weighted_f1":     rep["weighted avg"]["f1-score"],
        "history": history, "report_dict": rep,
        "test_preds": te_preds, "test_labels": te_labels,
    }
    for cls in emotion_classes:
        result[f"f1_{cls}"] = rep[cls]["f1-score"] if cls in rep else 0.0
    return result

# ============================================================
# ЗАГРУЗКА LLM-АУГМЕНТИРОВАННОГО ТРЕЙНА
# ============================================================

def load_llm_augmented(base_train_df):
    """
    Загружает train_augmented.csv.
    Пересчитывает label_id из final_label (фиксированный маппинг).
    Проверяет что val/test примеры не попали в аугментированный train.
    """
    path = os.path.join(AUG_DIR, "train_augmented.csv")
    df   = pd.read_csv(path)

    # Фиксируем label_id
    df["label_id"] = df["final_label"].map(CLASS_TO_ID)
    assert df["label_id"].isna().sum() == 0, "Unmapped labels!"

    # Проверка на утечку — ищем оригинальные тексты (не аугментированные)
    # которых нет в base_train_df
    if "is_augmented" in df.columns:
        original_in_aug  = set(df[~df["is_augmented"]]["tweet_text"].str.strip().str.lower())
        original_in_base = set(base_train_df["tweet_text"].str.strip().str.lower())
        leaked = original_in_aug - original_in_base
        if leaked:
            print(f"  ⚠️  WARNING: {len(leaked)} original texts not found in base train!")
            print(f"     These may have leaked from old split. Examples:")
            for t in list(leaked)[:3]:
                print(f"       '{t[:80]}'")
        else:
            print(f"  ✅ No leakage detected — all original texts match base train")

    print(f"\n📥 LLM-augmented train: {len(df)} samples")
    for cls in emotion_classes:
        total = (df["final_label"] == cls).sum()
        aug   = 0
        if "is_augmented" in df.columns:
            aug = ((df["final_label"] == cls) & df["is_augmented"]).sum()
        print(f"  id={CLASS_TO_ID[cls]}  {cls:<12}: {total:>5}  (aug={aug})")

    return df

# ============================================================
# ЗАГРУЗКА СПЛИТОВ И ЗАПУСК
# ============================================================

loaders, meta = load_splits(splits_dir=SPLITS_DIR)
train_df  = meta["train_df"]
val_df    = meta["val_df"]
test_df   = meta["test_df"]
tokenizer = meta["tokenizer"]

print(f"\nVal size:  {len(val_df)}")
print(f"Test size: {len(test_df)}")

# ============================================================
# ТОЛЬКО LLM AUGMENTATION
# ============================================================

train_llm = load_llm_augmented(base_train_df=train_df)

res_llm = run_experiment(
    train_llm, val_df, test_df,
    display_name="LLM Augmentation (DeepSeek)",
    tokenizer=tokenizer,
)

# ============================================================
# ГРАФИКИ
# ============================================================

plt.rcParams['font.family'] = 'DejaVu Sans'
PROPS_BOX = dict(boxstyle='round,pad=0.6', facecolor='#f5f5f5', edgecolor='#aaaaaa', alpha=0.9)


def plot_training_curves(result, output_dir):
    ep = range(1, len(result["history"]["train_f1"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.patch.set_facecolor('white')
    for ax in axes:
        ax.set_facecolor('white')
        ax.grid(True, ls='--', alpha=0.4, zorder=1)
        ax.set_axisbelow(True)
    axes[0].plot(ep, result["history"]["train_f1"], "o-", color="steelblue", lw=2.5, ms=6, label="Train F1", zorder=3)
    axes[0].plot(ep, result["history"]["val_f1"],   "s-", color="#e74c3c",   lw=2.5, ms=6, label="Val F1",   zorder=3)
    axes[0].set_title("Macro F1", fontsize=13, fontweight="bold")
    axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Macro F1"); axes[0].legend(fontsize=11)
    axes[1].plot(ep, result["history"]["train_loss"], "o-", color="steelblue", lw=2.5, ms=6, label="Train Loss", zorder=3)
    axes[1].plot(ep, result["history"]["val_loss"],   "s-", color="#e74c3c",   lw=2.5, ms=6, label="Val Loss",   zorder=3)
    axes[1].set_title("Loss", fontsize=13, fontweight="bold")
    axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Loss"); axes[1].legend(fontsize=11)

    stats = (f"Epochs: {result['epochs_trained']}  |  "
             f"Best Val F1: {result['best_val_f1']:.4f}  |  "
             f"Train N: {result['train_size']:,}")
    axes[1].text(0.99, 0.03, stats, transform=axes[1].transAxes, fontsize=9,
                 va='bottom', ha='right', bbox=PROPS_BOX, family='monospace')

    plt.suptitle(result["variant"], fontsize=13, fontweight="bold", y=1.01)
    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"curves_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_per_class_f1(result, output_dir):
    rep = result["report_dict"]
    f1s = [rep[c]["f1-score"] if c in rep else 0 for c in emotion_classes]
    fig, ax = plt.subplots(figsize=(11, 6))
    fig.patch.set_facecolor('white'); ax.set_facecolor('white')
    bars = ax.bar(emotion_classes, f1s, color="steelblue", alpha=0.75, edgecolor='white', zorder=2)
    for b, v in zip(bars, f1s):
        ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.012,
                f"{v:.4f}", ha="center", fontsize=10, fontweight="bold")
    ax.axhline(result["test_macro_f1"], color="#FFA500", ls="--", lw=2.2, zorder=3,
               label=f"Macro F1={result['test_macro_f1']:.4f}")
    ax.set_ylim(0, 1.05); ax.set_ylabel("F1"); ax.set_xlabel("Class")
    ax.set_title(f"Per-Class F1 — {result['variant']}", fontsize=12, fontweight="bold", pad=12)
    ax.grid(True, axis='y', ls='--', alpha=0.4, zorder=1); ax.set_axisbelow(True); ax.legend(fontsize=11)

    stats = (f"Acc={result['test_accuracy']:.4f}  "
             f"P={result['test_macro_precision']:.4f}  "
             f"R={result['test_macro_recall']:.4f}  "
             f"F1={result['test_macro_f1']:.4f}  "
             f"wF1={result['test_weighted_f1']:.4f}")
    ax.text(0.99, 0.03, stats, transform=ax.transAxes, fontsize=9,
            va='bottom', ha='right', bbox=PROPS_BOX, family='monospace')

    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"perclass_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


def plot_confusion_matrix(result, output_dir):
    fig, ax = plt.subplots(figsize=(9, 7))
    fig.patch.set_facecolor('white')
    cm   = confusion_matrix(result["test_labels"], result["test_preds"])
    cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)
    sns.heatmap(cm_n, annot=True, fmt=".2f", cmap="Blues",
                xticklabels=emotion_classes, yticklabels=emotion_classes,
                ax=ax, linewidths=0.5, linecolor='white')
    ax.set_title(f"CM — {result['variant']}\n"
                 f"F1={result['test_macro_f1']:.4f}  Acc={result['test_accuracy']:.4f}",
                 fontsize=11, fontweight="bold", pad=12)
    ax.set_ylabel("True"); ax.set_xlabel("Predicted")
    plt.tight_layout()
    safe = re.sub(r'[^\w\-]', '_', result['variant'])[:40]
    plt.savefig(os.path.join(output_dir, f"cm_{safe}.png"),
                dpi=150, bbox_inches="tight", facecolor='white')
    plt.show()


plot_training_curves(res_llm, OUTPUT_DIR)
plot_per_class_f1(res_llm, OUTPUT_DIR)
plot_confusion_matrix(res_llm, OUTPUT_DIR)

# ============================================================
# РЕЗУЛЬТАТЫ
# ============================================================

print(f"\n{'='*70}")
print("  LLM AUGMENTATION RESULT")
print(f"{'='*70}\n")
print(f"  Train size:   {res_llm['train_size']:,}")
print(f"  Epochs:       {res_llm['epochs_trained']}")
print(f"  Accuracy:     {res_llm['test_accuracy']:.4f}")
print(f"  Macro P:      {res_llm['test_macro_precision']:.4f}")
print(f"  Macro R:      {res_llm['test_macro_recall']:.4f}")
print(f"  Macro F1:     {res_llm['test_macro_f1']:.4f}")
print(f"  Weighted F1:  {res_llm['test_weighted_f1']:.4f}")
print(f"  Best Val F1:  {res_llm['best_val_f1']:.4f}")

print(f"\n  Per-Class F1:")
for cls in emotion_classes:
    f1  = res_llm.get(f"f1_{cls}", 0)
    mrk = " ← minority" if cls in MINORITY_CLASSES else ""
    print(f"    {cls:<12}: {f1:.4f}{mrk}")

# Сохраняем одиночный результат
single_row = {
    "Method":    res_llm["variant"],
    "Train N":   res_llm["train_size"],
    "Epochs":    res_llm["epochs_trained"],
    "Accuracy":  round(res_llm["test_accuracy"], 4),
    "Macro P":   round(res_llm["test_macro_precision"], 4),
    "Macro R":   round(res_llm["test_macro_recall"], 4),
    "Macro F1":  round(res_llm["test_macro_f1"], 4),
    "wF1":       round(res_llm["test_weighted_f1"], 4),
    "Val F1":    round(res_llm["best_val_f1"], 4),
    **{f"f1_{cls}": round(res_llm.get(f"f1_{cls}", 0), 4) for cls in emotion_classes},
}
pd.DataFrame([single_row]).to_csv(
    os.path.join(OUTPUT_DIR, "llm_aug_result.csv"), index=False)

print(f"\nSaved to: {OUTPUT_DIR}/llm_aug_result.csv")
```
